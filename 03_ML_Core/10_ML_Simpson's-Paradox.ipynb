# Topic: ML: Simpson's Paradox

## Definition (30-second explanation)
Simpson's Paradox occurs when a trend or association observed in aggregated data completely reverses or disappears when that same data is separated into subgroups. It proves that correlation does not equal causation.

## Why Interviewers Ask This
*   **A/B Testing Pitfalls:** It is the primary reason poorly randomized A/B tests fail and lead to disastrous product decisions.
*   **Causal Inference:** Tests if you naturally think about "lurking" or confounding variables rather than blindly trusting aggregate metrics.
*   **Data Skepticism:** Interviewers want to hear the phrase: *"I always segment my data before drawing conclusions from aggregates."*

## Core Concepts
*   **Confounding Variable:** A hidden variable that influences both the independent variable (treatment/group) and the dependent variable (outcome).
*   **Unequal Distributions:** The paradox only happens when the subgroups have heavily skewed or unequal sample sizes across the treatments.
*   **Stratification:** The process of dividing data into homogenous subgroups (strata) to control for confounders.

## When to Use
*   Analyzing A/B test results (checking for sample ratio mismatch or unequal segmentation).
*   Evaluating medical trials or observational data where proper randomization wasn't possible.
*   Investigating any highly counterintuitive aggregate metric (e.g., "Why did our overall conversion rate drop when conversion went up on both iOS and Android?").

## Advantages
*   **Uncovers the Truth:** Properly stratifying data reveals the true causal direction of a feature or treatment.
*   **Improves Experimentation:** Forcing analysts to check for confounders leads to better A/B test design and rigorous randomization.

## Limitations
*   **The "Turtles All The Way Down" Problem:** You can theoretically keep slicing data into infinitely smaller subgroups (by age, then by device, then by location) until sample sizes are too small to be statistically significant. You must rely on domain knowledge to pick the *right* confounders.

## Common Comparisons
*   **Observational Data vs. RCTs:** Simpson's Paradox is highly common in observational data. Randomized Controlled Trials (RCTs) defeat the paradox because true randomization evenly distributes confounding variables across both groups.
*   **Aggregate vs. Segmented:** Aggregate shows "what" is happening overall; segmented shows "why" it is happening.

## Common Interview Traps
*   **Trusting the Aggregate:** Looking at a dashboard's top-line metric and declaring a winner without asking, "Are the treatment groups balanced?"
*   **Ignoring Sample Sizes:** Failing to notice that Treatment A was given to 1,000 easy cases and Treatment B was given to 10 hard cases.

## Important Formula / Math Proof
Aggregate tells a lie if:
*   $\frac{a_1}{b_1} > \frac{c_1}{d_1}$ (Treatment A wins Subgroup 1)
*   $\frac{a_2}{b_2} > \frac{c_2}{d_2}$ (Treatment A wins Subgroup 2)
*   BUT: $\frac{a_1 + a_2}{b_1 + b_2} < \frac{c_1 + c_2}{d_1 + d_2}$ (Treatment B wins overall)
