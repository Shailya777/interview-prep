# Topic: ML: Simpson's Paradox

## Definition (30-second explanation)
Simpson's Paradox occurs when a trend or association observed in aggregated data completely reverses or disappears when that same data is separated into subgroups. It proves that correlation does not equal causation.

## Why Interviewers Ask This
*   **A/B Testing Pitfalls:** It is the primary reason poorly randomized A/B tests fail and lead to disastrous product decisions.
*   **Causal Inference:** Tests if you naturally think about "lurking" or confounding variables rather than blindly trusting aggregate metrics.
*   **Data Skepticism:** Interviewers want to hear the phrase: *"I always segment my data before drawing conclusions from aggregates."*

## Core Concepts
*   **Confounding Variable:** A hidden variable that influences both the independent variable (treatment/group) and the dependent variable (outcome).
*   **Unequal Distributions:** The paradox only happens when the subgroups have heavily skewed or unequal sample sizes across the treatments.
*   **Stratification:** The process of dividing data into homogenous subgroups (strata) to control for confounders.

## When to Use
*   Analyzing A/B test results (checking for sample ratio mismatch or unequal segmentation).
*   Evaluating medical trials or observational data where proper randomization wasn't possible.
*   Investigating any highly counterintuitive aggregate metric (e.g., "Why did our overall conversion rate drop when conversion went up on both iOS and Android?").

## Advantages
*   **Uncovers the Truth:** Properly stratifying data reveals the true causal direction of a feature or treatment.
*   **Improves Experimentation:** Forcing analysts to check for confounders leads to better A/B test design and rigorous randomization.

## Limitations
*   **The "Turtles All The Way Down" Problem:** You can theoretically keep slicing data into infinitely smaller subgroups (by age, then by device, then by location) until sample sizes are too small to be statistically significant. You must rely on domain knowledge to pick the *right* confounders.

## Common Comparisons
*   **Observational Data vs. RCTs:** Simpson's Paradox is highly common in observational data. Randomized Controlled Trials (RCTs) defeat the paradox because true randomization evenly distributes confounding variables across both groups.
*   **Aggregate vs. Segmented:** Aggregate shows "what" is happening overall; segmented shows "why" it is happening.

## Common Interview Traps
*   **Trusting the Aggregate:** Looking at a dashboard's top-line metric and declaring a winner without asking, "Are the treatment groups balanced?"
*   **Ignoring Sample Sizes:** Failing to notice that Treatment A was given to 1,000 easy cases and Treatment B was given to 10 hard cases.

## Important Formula / Math Proof
Aggregate tells a lie if:
*   $\frac{a_1}{b_1} > \frac{c_1}{d_1}$ (Treatment A wins Subgroup 1)
*   $\frac{a_2}{b_2} > \frac{c_2}{d_2}$ (Treatment A wins Subgroup 2)
*   BUT: $\frac{a_1 + a_2}{b_1 + b_2} < \frac{c_1 + c_2}{d_1 + d_2}$ (Treatment B wins overall)

## 45-Second Interview Answer
"Simpson's Paradox happens when a trend that is obvious in aggregated data completely reverses when you segment the data into subgroups. This is almost always caused by a confounding variable combined with an unequal distribution of data—for example, if Treatment B in an A/B test was accidentally shown mostly to desktop users who naturally convert at higher rates. To prevent it, I always ensure A/B tests are properly randomized, and I make it a rule to segment observational data by key dimensions before reporting top-line conclusions."

## Example Questions:

### Q1. Hospital A has a higher overall survival rate than Hospital B, but Hospital B has better survival rates for both minor and major surgeries. Explain why and which hospital you would choose.

**Ideal Answer:**
"This is a classic example of Simpson's Paradox. I would choose Hospital B. The paradox is occurring because 'Surgery Severity' is a confounding variable. Hospital A likely handles mostly minor, low-risk surgeries, inflating its overall survival rate. Hospital B acts as a trauma center handling the most severe, high-risk cases, which drags down its overall average. However, when comparing apples-to-apples (minor to minor, major to major), Hospital B performs better. You must look at the stratified data to make the correct choice."

*   **Common Mistakes:** Attempting to average the percentages of the subgroups without weighting them by the number of patients, or picking Hospital A just because the top-line number is higher.
*   **Interviewer Follow-up:** "How would you statistically prove this to a non-technical hospital administrator?" (Answer: I would show them a simple bar chart of the patient distribution, visually demonstrating that Hospital A gets 90% easy cases while Hospital B gets 90% hard cases).


### Q2. Real Interview Q (Airbnb): Our A/B test shows that Version B has higher conversion overall. But our PM says B is worse. Who is right?

**Ideal Answer:**
"Both could be right, which immediately points to Simpson's Paradox. If the A/B test wasn't properly randomized, a lurking variable could be skewing the aggregate. My investigation would be to segment the data by key dimensions—like device type, country, or traffic source. I would calculate the conversion rates for A and B *within* each segment, and then check the traffic distribution. If Version B was disproportionately shown to historically high-converting segments (like returning desktop users), it would artificially inflate B's overall score, making the PM's segmented view the correct one."

*   **Common Mistakes:** Assuming the PM is wrong because "the data doesn't lie," failing to recognize that A/B tests can have flawed randomization setups.
*   **Interviewer Follow-up:** "If you find this happened, how do you fix the A/B test?" (Answer: We need to halt the test, fix the routing logic to ensure strict randomization so both variants get identical distributions of user types, and restart the test).


### Q3. Explain Simpson's Paradox to a non-technical stakeholder using a sports analogy.

**Ideal Answer:**
"Imagine two baseball players, Alice and Bob. In 2024, Alice had a better batting average than Bob. In 2025, Alice again had a better batting average than Bob. But if you combine their hits over *both* years, Bob actually has the higher overall average. 
This happens because of the number of times they went to bat. Bob might have had 500 at-bats in his good year and only 10 in his bad year, while Alice had 500 at-bats in her bad year and only 10 in her good year. The sheer volume of Bob's good year overpowers the math when you lump it all together. It shows why we have to look at the context, not just the final total."

*   **Common Mistakes:** Using overly complex statistical jargon (like "confounders" or "covariates") when speaking to a non-technical stakeholder.
*   **Interviewer Follow-up:** "How does this sports analogy translate directly to our marketing campaigns?" (Answer: A campaign might look like a failure overall, but if we realize it was mostly targeted at a notoriously difficult demographic, it might actually be outperforming expectations within that specific group).
