# Topic: ML: Train-Test Split

## Definition (30-second explanation)
* The Train-Test Split separates a dataset into a **training set** (to teach the model) and a **test set** (to evaluate generalization to unseen data).
* It ensures you do not measure a model's performance on data it has already seen, which would yield misleading, overly optimistic results.
* It is a fundamental safeguard against **data leakage** and model overfitting.

## Why Interviewers Ask This
* **Foundation Check:** It tests if you understand the core purpose of machine learning: generalization to *future/unseen* data, not just memorization.
* **Practical Rigor:** Interviewers look for professional habits, specifically using `random_state` for reproducibility and `stratify` for classification tasks.
* **Leakage Awareness:** It evaluates your ability to spot scenarios where a naive random split would cause data leakage (e.g., time-series data).

## Core Concepts
* **Data Leakage:** Occurs if the model is trained and evaluated on the same data, causing it to memorize patterns rather than learn generalizable features.
* **Stratification (`stratify=y`):** Ensures the distribution of target classes remains identical in both the training and test sets, which is critical for imbalanced data.
* **Reproducibility (`random_state`):** Setting a random seed ensures that the exact same random split occurs every time the code is run, enabling consistent debugging and team collaboration.

## When to Use
* When building any supervised machine learning model (classification or regression).
* When simulating how a model will behave on real-world, future data.
* When the dataset is sufficiently large that a single hold-out set provides a reliable performance estimate.

## Advantages
* Provides a quick, simple, and computationally cheap validation approach.
* Offers an objective evaluation of model performance on completely unseen data.

## Limitations
* Unreliable for very small datasets (fewer than 500 rows); the test set may not be fully representative of the data distribution.
* A naive random split will fail on data with a time-series structure.

## Common Comparisons
* **Train-Test Split vs. K-Fold Cross-Validation:** Train-test split is faster and simpler but riskier on small data. K-Fold CV splits data multiple times to ensure highly reliable performance estimates, preferred for small datasets.
* **Two-Way vs. Three-Way Split:** A two-way split (Train/Test) evaluates final performance. A three-way split (Train/Validation/Test) is used when tuning hyperparameters, using the validation set for tuning without contaminating the final test set.

## Common Interview Traps
* **Evaluating on Train Data:** Reporting 'test accuracy' based on training data evaluation, artificially inflating the metric.
* **Ignoring Imbalance:** Forgetting to use `stratify=y` on classification problems, resulting in a test set with 0 positive examples.
* **Random Splitting Time-Series:** Using `train_test_split` on temporal data, which leaks future information into the training past.
