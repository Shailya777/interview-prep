### 1. Confusion Matrix
| | Predicted: 0 | Predicted: 1 |
| :--- | :--- | :--- |
| **Actual: 0** | TN | FP |
| **Actual: 1** | FN | TP |

### 2. Accuracy
Truly predicted points divided by everything.
* `(TP + TN) / (TP + FP + TN + FN)`

### 3. Precision
From what the model predicted as positive, how many were actually positive. All about predicted positives. 
* `TP / (TP + FP)`

### 4. Recall
From all actual positives, how many did the model catch correctly. All about Actual Positives. 
* `TP / (TP + FN)`

### 5. TPR (True Positive Rate)
Is recall, sensitivity.

### 6. FPR (False Positive Rate)
1 - Specificity (just change all T and F from recall formula).
* `FP / (FP + TN)`

### 7. F1-Score
Harmonic, not arithmetic mean of Precision and Recall. Arithmetic just averages. Harmonic makes sure if gap between two is too high, score drops.
* `2 * (Precision * Recall) / (Precision + Recall)`

**F-beta Score:** If we want to weight either of Precision or Recall more, we can use fbeta score. 
* **Beta < 1:** Makes precision more important, getting false positives costlier like in spam detection. 
* **Beta > 1:** Makes recall more important, making false negatives costlier, like in Disease detection. 

**Multi-class F1:**
* **Macro F1:** Calculates F1 for all classes independently, giving equal importance to minority class.
* **Micro F1:** Calculates global TP, FP, FN, which can skew results in case of imbalanced data.
* **Weighted F1:** Takes actual data points, support, in each class to decide how much weight to give each class, again making minority class in imbalanced data less important.

### 8. ROC-AUC
Threshold independent way of measuring classifier performance. AUC shows how much probability model has of giving higher predicted probability to random positive point than random negative point. 

* Plots True Positive Rate vs False Positive Rate. 
* Max of `tpr - fpr` shows which threshold is best. 
* Does not work well with Imbalanced data as True negatives are in the denominator of FPR, which can make FPR **artificially low**. 
* For imbalanced data, **PR-AUC** (Precision-Recall AUC) is better, which plots precision vs recall.
