# Topic: Poisson Distribution

## Definition (30-second explanation)
The Poisson distribution models the probability of a specific number of events occurring within a fixed interval of time or space. It is defined entirely by a single parameter, $\lambda$ (lambda), which represents the average rate of occurrence, assuming these events are rare and occur independently.

## Why Interviewers Ask This
Interviewers use this to test your understanding of probability distributions applied to real-world operations, such as queueing theory, server load estimation, customer arrivals, and identifying anomalies (rare events).

## Core Concepts
* **Single Parameter ($\lambda$):** The average number of events per interval dictates the entire distribution shape.
* **Unique Property:** Mean = Variance = $\lambda$. 
* **Independence:** The occurrence of one event does not affect the probability of another.
* **Asymptotic Behavior:** As $\lambda$ increases, the Poisson distribution approximates a Normal distribution. It is also the limit of the Binomial distribution when $n$ is very large and $p$ is very small ($\lambda = np$).

## When to Use
* Counting events in a fixed time/space boundary.
* Examples: Server API requests per minute, customer arrivals per hour, network packet drops, or rare defects in manufacturing.

## Advantages
* Mathematically elegant and requires only historical averages ($\lambda$) to predict future probabilities.
* Easily implemented using standard statistical libraries.

## Limitations
* **Strict Variance Assumption:** Real-world data often violates the Mean = Variance rule (commonly showing "overdispersion" where Variance > Mean).
* **Strict Independence:** Cannot be used if events trigger cascading effects (e.g., viral social media shares).

## Common Comparisons
* **vs. Binomial:** Binomial counts successes in a fixed number of trials ($n$); Poisson counts events in a continuous interval where $n \to \infty$.
* **vs. Negative Binomial:** Used instead of Poisson when data is overdispersed (Variance > Mean).

## Common Interview Traps
* **Forgetting the Variance Rule:** Candidates often fail to mention that the mean and variance must be equal for a valid Poisson model.
* **Using for Dependent Events:** Applying it to scenarios like contagious diseases where one event increases the likelihood of the next.

## Python / SQL Syntax
```python
from scipy import stats
import numpy as np

lambda_rate = 8

# P(X = 5) - Exactly 5 events (Probability Mass Function)
prob_exact = stats.poisson.pmf(5, lambda_rate)

# P(X <= 9) - Cumulative Distribution Function
prob_cumulative = stats.poisson.cdf(9, lambda_rate)

# P(X >= 10) - Survival Function (1 - CDF)
prob_greater_equal = 1 - stats.poisson.cdf(9, lambda_rate)

# Generate simulated data
simulated_data = np.random.poisson(lambda_rate, size=1000)
