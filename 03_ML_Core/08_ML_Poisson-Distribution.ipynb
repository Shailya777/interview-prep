# Topic: Poisson Distribution

## Definition (30-second explanation)
The Poisson distribution models the probability of a specific number of events occurring within a fixed interval of time or space. It is defined entirely by a single parameter, $\lambda$ (lambda), which represents the average rate of occurrence, assuming these events are rare and occur independently.

## Why Interviewers Ask This
Interviewers use this to test your understanding of probability distributions applied to real-world operations, such as queueing theory, server load estimation, customer arrivals, and identifying anomalies (rare events).

## Core Concepts
* **Single Parameter ($\lambda$):** The average number of events per interval dictates the entire distribution shape.
* **Unique Property:** Mean = Variance = $\lambda$. 
* **Independence:** The occurrence of one event does not affect the probability of another.
* **Asymptotic Behavior:** As $\lambda$ increases, the Poisson distribution approximates a Normal distribution. It is also the limit of the Binomial distribution when $n$ is very large and $p$ is very small ($\lambda = np$).

## When to Use
* Counting events in a fixed time/space boundary.
* Examples: Server API requests per minute, customer arrivals per hour, network packet drops, or rare defects in manufacturing.

## Advantages
* Mathematically elegant and requires only historical averages ($\lambda$) to predict future probabilities.
* Easily implemented using standard statistical libraries.

## Limitations
* **Strict Variance Assumption:** Real-world data often violates the Mean = Variance rule (commonly showing "overdispersion" where Variance > Mean).
* **Strict Independence:** Cannot be used if events trigger cascading effects (e.g., viral social media shares).

## Common Comparisons
* **vs. Binomial:** Binomial counts successes in a fixed number of trials ($n$); Poisson counts events in a continuous interval where $n \to \infty$.
* **vs. Negative Binomial:** Used instead of Poisson when data is overdispersed (Variance > Mean).

## Common Interview Traps
* **Forgetting the Variance Rule:** Candidates often fail to mention that the mean and variance must be equal for a valid Poisson model.
* **Using for Dependent Events:** Applying it to scenarios like contagious diseases where one event increases the likelihood of the next.

## Python / SQL Syntax
```python
from scipy import stats
import numpy as np

lambda_rate = 8

# P(X = 5) - Exactly 5 events (Probability Mass Function)
prob_exact = stats.poisson.pmf(5, lambda_rate)

# P(X <= 9) - Cumulative Distribution Function
prob_cumulative = stats.poisson.cdf(9, lambda_rate)

# P(X >= 10) - Survival Function (1 - CDF)
prob_greater_equal = 1 - stats.poisson.cdf(9, lambda_rate)

# Generate simulated data
simulated_data = np.random.poisson(lambda_rate, size=1000)
```

## Important Formula
Probability Mass Function:
$$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$$
Mean and Variance:
$$E(X) = Var(X) = \lambda$$

## 45-Second Interview Answer
"The Poisson distribution models the count of independent events occurring in a fixed interval, like server requests per minute. Its defining characteristic is that both its mean and variance are equal to lambda ($\lambda$), the average event rate. I use it to estimate probabilities of system overloads or rare anomalies, but I always check for overdispersion first, as real-world data often has a variance higher than its mean, which would require a Negative Binomial model instead."

### Example Practice Question 1: Customer Service Queue
**Question:** A customer service center receives an average of 8 calls per hour. 
(a) What is the probability of receiving exactly 5 calls in the next hour? 
(b) What is the probability of receiving 10 or more calls in the next hour?

**Ideal Interview Answer:** 
"Since this is a count of independent events in a fixed time interval, we model it with a Poisson distribution where $\lambda = 8$. 
For part (a), finding exactly 5 calls requires the Probability Mass Function (PMF). I'd calculate $\frac{8^5 e^{-8}}{5!}$, which equals about 9.16%.
For part (b), finding 10 or more calls requires $1 - P(X \le 9)$. I would use the Cumulative Distribution Function (CDF) for $X=9$ and subtract it from 1, yielding approximately 28.34%."

**Common Mistakes:** 
* For part (b), calculating $1 - P(X \le 10)$ instead of $1 - P(X \le 9)$. "10 or more" means 10 is included in the desired outcome, so it must be excluded from the CDF subtraction.

**Likely Follow-up:** 
"If the calls suddenly started coming in waves because of a system outage, would this model still be accurate?"

### Example Practice Question 2: Server Overload Modeling
**Question:** A website gets an average of 100 visits per minute. The server overloads if capacity exceeds 120 visits per minute. How do you calculate the probability of a server overload?

**Ideal Interview Answer:**
"We set our rate parameter $\lambda = 100$. An overload occurs when visits are strictly greater than 120. To find this probability, I would calculate $1 - P(X \le 120)$ using the Poisson CDF. In Python, this is `1 - scipy.stats.poisson.cdf(120, 100)`. Because $\lambda$ is large (100), we could also approximate this using a Normal distribution with a mean of 100 and a standard deviation of 10 to check our bounds quickly."

**Common Mistakes:**
* Not recognizing that for high values of $\lambda$, the distribution approaches normality, which is a great optimization to mention in an interview.

**Likely Follow-up:**
"How would you simulate a month of this traffic to test your auto-scaling architecture?"

## Practice Questions:

### Q1:
**You are deploying a new Natural Language Processing API. You observe that the API receives an average of 50 requests per minute ($\lambda = 50$). You plan to use a Poisson distribution to predict traffic spikes and configure your cloud auto-scaling rules.However, during exploratory data analysis, you calculate the actual variance of your historical requests per minute, and you find the variance is 150.Would you still proceed with using the Poisson distribution for your auto-scaling model? Why or why not? If not, what distribution might you use instead?**

**Answer:**
"No, I would not proceed with the Poisson distribution. A core assumption of the Poisson model is that the mean and variance must be equal. Here, the variance (150) is three times the mean (50), which is a condition called overdispersion. This typically happens with API traffic because requests aren't strictly independent—they often come in bursts. Because the variance is strictly greater than the mean, I would use a Negative Binomial distribution instead to model the counts accurately. Alternatively, since our lambda is relatively large (50), I could also explore a Normal distribution approximation depending on the strictness of the discrete count requirement."

**Common Mistakes:**
* Ignoring the variance entirely and just blindly plugging $\lambda=50$ into a Poisson function.
* Suggesting the Binomial distribution, which is incorrect because we do not have a fixed number of trials ($n$).

**Likely Follow-up:**
"Write a quick Python script to check for overdispersion in a dataset."
