# Topic: ML Label Encoding & Ordinal Encoding

## Definition (30-second explanation)
*   **Label Encoding** converts each unique category in a column into a unique integer (e.g., Red=0, Blue=1, Green=2).
*   It is memory-efficient and avoids dimensionality expansion but implies a mathematical ordering where none may exist.

## Why Interviewers Ask This
*   It tests your understanding of how different algorithms interpret numerical inputs (gradients/weights vs. binary splits).
*   It checks if you know the critical API difference in Scikit-Learn between encoding targets (`y`) and encoding features (`X`).

## Core Concepts
*   **Artificial Ordering:** Assigning 0, 1, 2 to nominal data implies that 2 is "greater" than 1, which confuses linear models.
*   **Alphabetical Default:** By default, simple label encoders assign integers based on alphabetical order, which can destroy the logic of actual ordinal data (e.g., 'Low', 'Medium', 'High').
*   **Dimensionality Preservation:** Unlike One-Hot Encoding, Label Encoding keeps the data in a single column.

## When to Use
*   **Target Variables ($y$):** Standard practice for multi-class classification targets.
*   **Ordinal Variables:** When categories have a natural ranking (e.g., Education Level), provided you map them in the correct order.
*   **Tree-based Models:** Decision Trees, Random Forests, and XGBoost can partition integer-mapped categories without assuming linear relationships.

## Advantages
*   Zero increase in dimensionality, keeping memory footprint very low.
*   Highly efficient for tree-based models compared to massive, sparse One-Hot Encoded matrices.
*   Simple to implement for binary variables (though manual 0/1 mapping is often clearer).

## Limitations
*   Disastrous for Linear Regression, Logistic Regression, SVMs, or K-Nearest Neighbors if used on nominal data.
*   Can still cause inefficient, deep splitting in standard Decision Trees if applied to high-cardinality nominal features.

## Common Comparisons
*   **LabelEncoder vs. OrdinalEncoder:** `LabelEncoder` is strictly for 1D target arrays ($y$) and assigns alphabetically. `OrdinalEncoder` is for 2D feature matrices ($X$) and allows passing a specific ordered list to enforce the correct logical hierarchy.
*   **Label Encoding vs. One-Hot Encoding:** Use Label/Ordinal for ordered data or trees; use One-Hot for nominal data and linear/distance-based models.

## Common Interview Traps
*   **The Alphabetical Trap:** Blindly applying `LabelEncoder` to an ordinal feature like `['Small', 'Medium', 'Large']`. It will encode them as Large=0, Medium=1, Small=2, completely ruining the pattern.
*   **The Linear Trap:** Using it for nominal features (like "City") in a linear model, tricking the model into learning that "Tokyo" is mathematically greater than "London".
