# Topic: Precision vs Recall (and Classification Tradeoffs)

## Definition (30-second explanation)
* **Precision:** Out of all the positive predictions the model made, how many were actually correct? (Focuses on minimizing False Positives).
* **Recall (Sensitivity):** Out of all the actual positive cases in the data, how many did the model successfully find? (Focuses on minimizing False Negatives).
* **The Tradeoff:** Governed by the classification threshold. Increasing the threshold raises precision but lowers recall, and vice-versa. 

## Why Interviewers Ask This
* Evaluates your ability to translate real-world business costs (e.g., the cost of a missed cancer diagnosis vs. a blocked legitimate transaction) into appropriate ML metrics.
* Tests your understanding of why standard "Accuracy" is severely misleading on imbalanced datasets.

## Core Concepts
* **True Positives (TP) / True Negatives (TN):** Correct predictions.
* **False Positives (FP - Type I Error):** Model falsely flags a negative as positive (e.g., healthy patient diagnosed as sick).
* **False Negatives (FN - Type II Error):** Model misses an actual positive (e.g., sick patient diagnosed as healthy).
* **Threshold Tuning:** Adjusting the probability cutoff (default usually 0.5) to favor either precision or recall based on business needs.

## When to Use
* Whenever classes are imbalanced (e.g., 99% legitimate, 1% fraud).
* **High Precision required:** When False Positives are expensive or ruin user experience (Spam Filtering, Content Recommendations).
* **High Recall required:** When False Negatives are dangerous or cause severe losses (Cancer Detection, Security Breaches).
* **Balanced required:** Fraud detection, where both missed fraud (FN) and blocked good customers (FP) cost money (use F1-Score).

## Advantages
* Provides a granular, realistic view of model performance on minority classes.
* Directly maps to business KPIs (risk, cost, revenue) much better than global accuracy.

## Limitations
* They only focus on the performance of the positive class and completely ignore True Negatives (which Specificity captures).
* Precision can be artificially inflated by predicting only the single most confident positive instance (yielding 100% precision but near 0% recall).

## Common Comparisons
* **Precision vs Recall:** Precision penalizes false alarms; Recall penalizes missed detections.
* **Accuracy vs F1-Score:** Accuracy is for balanced data/equal error costs; F1-Score is the harmonic mean of precision and recall for imbalanced data.
* **Recall vs Specificity:** Recall measures accuracy on actual positives; Specificity measures accuracy on actual negatives.

## Common Interview Traps
* **Trap 1:** Reporting 99% accuracy on a dataset with 1% positive cases. (The model is likely just predicting the majority class).
* **Trap 2:** Confusing the denominators. Precision divides by *Predicted* Positives (TP + FP). Recall divides by *Actual* Positives (TP + FN).
* **Trap 3:** Claiming a model is perfect because it has 100% recall. (Predicting everything as positive gives 100% recall with terrible precision).
* **Trap 4:** Applying precision/recall to regression problems (they are classification metrics only).

## Python / SQL Syntax (if applicable)
* `from sklearn.metrics import precision_score, recall_score, confusion_matrix, classification_report`
* `precision = precision_score(y_true, y_pred)`
* Threshold adjustment: `y_pred_custom = (model.predict_proba(X_test)[:, 1] >= 0.7).astype(int)`

## Important Formula
* Precision = TP / (TP + FP)
* Recall = TP / (TP + FN)
* Accuracy = (TP + TN) / (TP + TN + FP + FN)

Recall answers: 'Of all the actual positives, how many did we catch?' It is critical when missing a case is dangerous, like in disease diagnosis. In practice, I tune the classification threshold to find the optimal tradeoff between the two based on the specific business objective, or use the F1-Score to balance them."

## PRACTICE QUESTIONS & INTERVIEW STRATEGIES

### Question 1
**Question:** A model has TP=50, TN=900, FP=10, FN=40. Calculate Precision, Recall, Accuracy, and F1 Score.
* **Ideal Answer:** 
  * Precision = TP / (TP + FP) = 50 / (50 + 10) = 50 / 60 = 0.833 (83.3%)
  * Recall = TP / (TP + FN) = 50 / (50 + 40) = 50 / 90 = 0.555 (55.5%)
  * Accuracy = (TP + TN) / Total = 950 / 1000 = 0.95 (95%)
  * F1 Score = 2 * (Precision * Recall) / (Precision + Recall) = 2 * (0.833 * 0.555) / (0.833 + 0.555) ≈ 0.666
* **Common Mistakes:** Messing up the denominators (e.g., using total population for precision instead of predicted positives).
* **Interviewer Follow-up:** Given these numbers, is this model better suited for a cancer screening task or a spam filtering task, and why?

### Question 2
**Question:** In a spam detection system, which metric would you prioritize and why? What threshold would you choose?
* **Ideal Answer:** I would prioritize Precision. The cost of a False Positive (sending a critical work email to the spam folder) is incredibly high and ruins user trust. The cost of a False Negative (a spam email reaches the inbox) is a minor annoyance. Therefore, I would choose a high classification threshold (e.g., 0.85 or 0.90) to ensure the model only flags an email as spam when it is highly confident, maximizing precision at the expense of recall.
* **Common Mistakes:** Prioritizing recall, which would aggressively filter emails and block legitimate messages.
* **Interviewer Follow-up:** If you raise the threshold to 0.99, what happens to your True Positives and False Negatives?

### Question 3
**Question:** Explain the precision-recall tradeoff. Why does increasing the classification threshold typically increase precision but decrease recall?
* **Ideal Answer:** Most classifiers output a probability between 0 and 1. If we increase the threshold (e.g., from 0.5 to 0.8), we force the model to be much more strict and confident before predicting a "1". Because it is only making "safe" bets, the proportion of correct positive predictions (Precision) goes up. However, because it is so strict, it will inevitably miss some borderline actual positive cases, turning them into False Negatives, which causes Recall to drop.
* **Common Mistakes:** Claiming that increasing the threshold *always* increases precision (precision can fluctuate slightly if the strictness eliminates true positives faster than false positives in small samples, though the general trend is upward).
* **Interviewer Follow-up:** How would you visualize this tradeoff to present it to a non-technical product manager?

### Question 4
**Question:** Your recall is 0.95 but precision is 0.10. What does this tell you about your model and how do you fix it?
* **Ideal Answer:** This tells me the model is aggressively over-predicting the positive class. It is catching 95% of the actual targets (high recall), but 90% of its positive predictions are false alarms (terrible precision). It is likely acting like a "broken alarm" that rings constantly. To fix it, I would increase the classification probability threshold to make the model stricter. I could also try re-weighting the classes, adjusting the decision boundary, or gathering more discriminative features to separate the classes better.
* **Common Mistakes:** Assuming the model is inherently terrible; in some extreme medical triage scenarios, this might actually be the intended baseline before secondary screening.
* **Interviewer Follow-up:** Without retraining the model, write the Python code logic to apply a stricter threshold to the model's existing predicted probabilities.

## Practice Questions:

### Q1:
You have trained a binary classification model. You extracted its predicted probabilities and merged them with the true labels into a Pandas DataFrame named df.

Using pure Pandas/NumPy (no sklearn.metrics), write a concise Python snippet to:
1. Apply a custom classification threshold of 0.60 (Any predicted_prob $\ge 0.60$ becomes a positive prediction, else negative).
2. Calculate and print the Precision and Recall based on this new threshold.

```python
import numpy as np
import pandas as pd

# Mock output from your classification model
np.random.seed(42)
data = {
    'actual_label': [0, 1, 0, 1, 1, 0, 0, 1, 0, 1],
    'predicted_prob': [0.10, 0.80, 0.40, 0.90, 0.30, 0.20, 0.60, 0.70, 0.50, 0.85]
}
df = pd.DataFrame(data)
```

```python
# Answer:

# 1. Apply threshold using vectorization
df['predicted_label'] = (df['predicted_prob'] >= 0.60).astype(int)

# 2. Calculate confusion matrix components using boolean masks (NO LOOPS)
tp = ((df['actual_label'] == 1) & (df['predicted_label'] == 1)).sum()
fp = ((df['actual_label'] == 0) & (df['predicted_label'] == 1)).sum()
fn = ((df['actual_label'] == 1) & (df['predicted_label'] == 0)).sum()

# 3. Calculate metrics with zero-division protection
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0

print(f"Precision: {precision}, Recall: {recall}")
```

- Interview Tips:

1. NEVER use .iterrows() or for loops on DataFrames in a coding interview unless absolutely forced to. Always use vectorized operations (boolean indexing like (df['col'] == X).sum()).

2. Protect your fractions: Always account for the possibility of dividing by zero when manually coding precision and recall, especially when testing extreme thresholds (like 0.99).
