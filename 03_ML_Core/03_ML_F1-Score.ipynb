# Topic: F1 Score and F-beta Variants

## Definition (30-second explanation)
* The **F1 Score** is the harmonic mean of Precision and Recall, providing a single metric that balances both. 
* It ranges from 0 (worst) to 1 (best).
* The **F-beta ($F_\beta$) Score** is a generalization that allows you to apply different weights to Precision or Recall based on business needs.

## Why Interviewers Ask This
* To verify you know how to evaluate models on imbalanced datasets where standard accuracy fails.
* To test your mathematical intuition: specifically, *why* we use a harmonic mean instead of a simple average (arithmetic mean) to penalize extreme imbalances between precision and recall.

## Core Concepts
* **Harmonic Mean:** Emphasizes lower values. If either Precision or Recall is close to 0, the F1 Score will be dragged down closer to 0.
* **Perfect F1 (1.0):** Requires *both* perfect precision and perfect recall.
* **F-beta variants:**
  * **F1 Score ($\beta=1$):** Precision and Recall are weighted equally.
  * **F2 Score ($\beta=2$):** Recall is considered twice as important as Precision.
  * **F0.5 Score ($\beta=0.5$):** Precision is considered twice as important as Recall.

## When to Use
* Imbalanced classification problems where True Negatives are not the primary focus.
* When you need a single, automated evaluation metric for hyperparameter tuning or model comparison.
* **F2:** Medical diagnosis, fraud detection (where missing a case is worse).
* **F0.5:** Spam filters, content recommendation (where false alarms are worse).

## Advantages
* Provides a robust, single-number summary of model performance for minority classes.
* Safely penalizes "dumb" models (e.g., predicting everything as positive gives high recall, but terrible precision, resulting in a poor F1).

## Limitations
* Completely ignores True Negatives (TN). If your business problem relies heavily on correctly identifying the negative class, ROC-AUC or Matthews Correlation Coefficient (MCC) might be better.
* Harder to explain to non-technical stakeholders than standard accuracy.

## Common Comparisons
* **Arithmetic Mean vs Harmonic Mean:** Arithmetic mean of P=1.0 and R=0.0 is 0.5 (misleadingly okay). Harmonic mean of P=1.0 and R=0.0 is 0.0 (accurately shows the model is useless).
* **Macro vs Micro vs Weighted F1 (Multi-class):** 
  * *Macro:* Calculates F1 for each class independently and averages them (treats all classes equally, good for minority classes).
  * *Micro:* Calculates global TP, FP, FN first (heavily influenced by majority classes).
  * *Weighted:* Averages class F1 scores based on their actual support/size.

## Common Interview Traps
* **Trap 1:** Averaging Precision and Recall linearly instead of using the formula.
* **Trap 2:** Defaulting to F1 without asking the interviewer about the business context. (Always ask if False Positives or False Negatives are more expensive before assuming $\beta=1$).
* **Trap 3:** Comparing F1 scores across datasets with vastly different baseline distributions.

## Python / SQL Syntax (if applicable)
* `from sklearn.metrics import f1_score, fbeta_score`
* Equal weight: `f1_score(y_true, y_pred)`
* Recall-heavy: `fbeta_score(y_true, y_pred, beta=2)`
* Multi-class: `f1_score(y_true, y_pred, average='macro')`

## Important Formula
* **F1 Score:** $F1 = 2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$
* **F-beta Score:** $F_\beta = (1 + \beta^2) \cdot \frac{\text{Precision} \cdot \text{Recall}}{(\beta^2 \cdot \text{Precision}) + \text{Recall}}$

## 45-Second Interview Answer
"The F1 score is the harmonic mean of precision and recall. It's the go-to evaluation metric for imbalanced classification tasks because accuracy is often misleading. We use a harmonic mean rather than a simple average because it strictly penalizes models that have extreme disparities—meaning a model must have solid performance in *both* precision and recall to get a high F1. If the business context dictates that missing a positive case is far worse, I would shift to an F2 score to weigh recall higher; if false alarms are worse, I'd use an F0.5 score."

## PRACTICE QUESTIONS & INTERVIEW STRATEGIES

### Question 1
**Question:** Two models are competing for a fraud detection task. Model A has Precision = 0.90, Recall = 0.40. Model B has Precision = 0.65, Recall = 0.85. Calculate F1 Scores and determine which model is better.
* **Ideal Answer:** 
  * Model A F1 = $2 \cdot (0.90 \cdot 0.40) / (0.90 + 0.40) = 0.72 / 1.30 = 0.554$
  * Model B F1 = $2 \cdot (0.65 \cdot 0.85) / (0.65 + 0.85) = 1.105 / 1.50 = 0.737$
  * Model B is the better model for this general task. F1 severely penalized Model A for having a very poor recall, demonstrating how the harmonic mean pulls the score down toward the lowest value. 
* **Common Mistakes:** Calculating the arithmetic mean ($[0.90 + 0.40]/2 = 0.65$ vs $[0.65 + 0.85]/2 = 0.75$), which fails to properly penalize Model A's low recall.
* **Interviewer Follow-up:** If the business decides that catching fraud is absolutely paramount and false alarms are an acceptable cost, how would you evaluate these models instead? (Answer: Use the F2 score to give recall twice the weight).
