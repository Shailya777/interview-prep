# Topic: ML | Binomial Distribution

## Definition (30-second explanation)
* The Binomial Distribution is a discrete probability distribution that models the number of successes in a fixed number of independent trials.
* Each trial must have exactly two possible outcomes (success or failure), and the probability of success must remain constant across all trials.

## Why Interviewers Ask This
* It is the mathematical engine behind conversion rate modeling and A/B testing.
* It tests your foundational understanding of discrete probability versus continuous probability.
* Interviewers want to see if you understand the strict assumptions required to use it (independence and constant probability) and when those assumptions fail in the real world.

## Core Concepts
* **Parameters:** Defined by $n$ (total number of trials) and $p$ (probability of success in a single trial).
* **PMF (Probability Mass Function):** Calculates the probability of getting *exactly* $k$ successes.
* **CDF (Cumulative Density Function):** Calculates the probability of getting *up to* $k$ successes (used for "at most" or "at least" questions).
* **Expected Value:** The mean number of successes is calculated simply as $n*p$.

## When to Use
* Modeling binary events: Click-through rates (clicked/didn't click), conversion rates (bought/didn't buy), or churn (canceled/renewed).
* Calculating sample sizes or p-values for standard A/B tests.
* Quality control (e.g., probability of finding exactly 3 defective widgets in a batch of 100).

## Advantages
* Highly interpretable and mathematically simple.
* Serves as a perfect baseline model before moving to more complex generative models.

## Limitations
* Assumes all trials are **100% independent**. (e.g., One user's behavior cannot influence another's).
* Assumes $p$ is **perfectly constant**. (e.g., Conversion rates don't drop at night or spike on weekends). 
* Can become computationally expensive for massively large factorials, requiring approximations.

## Common Comparisons
* **Binomial vs. Poisson:** Binomial has a fixed number of trials ($n$). Poisson is used when the number of trials is infinite or unknown, modeling events over a continuous interval of time or space. (Binomial approximates Poisson when $n$ is very large and $p$ is very small).
* **Binomial vs. Normal:** Binomial is discrete; Normal is continuous. For large $n$ (specifically when $np \ge 5$ and $n(1-p) \ge 5$), the Binomial distribution roughly forms a bell curve and approximates the Normal distribution.

## Common Interview Traps
* **CDF vs PMF Confusion:** Using `.pmf()` when the interviewer asks for the probability of finding "3 *or more* conversions" (which requires `1 - .cdf()`).
* **Ignoring the Combinatorics:** Forgetting to multiply by $\binom{n}{k}$ when calculating probabilities manually. You must account for the number of different ways the successes can be arranged.

## Important Formula (if applicable)
* **Probability Mass Function (PMF):** $P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}$
* **Mean ($\mu$):** $n * p$
* **Standard Deviation ($\sigma$):** $\sqrt{n * p * (1 - p)}$

## Example Questions:

### Q1. An e-commerce website has a 5% conversion rate. If 20 visitors come to the site today, what is the probability that exactly 2 make a purchase? What is the expected number of purchases and the standard deviation?

**Ideal Interview Answer:**
"First, we define our parameters: $n = 20$ trials, $p = 0.05$ success rate, and we want exactly $k = 2$ successes.
The expected number of purchases (the mean) is $n * p$, so $20 * 0.05 = 1$ purchase.
The standard deviation is $\sqrt{n * p * (1 - p)}$, which is $\sqrt{20 * 0.05 * 0.95} = \sqrt{0.95} \approx 0.97$.
To find the probability of exactly 2 purchases, we use the PMF. We calculate the combinations $\binom{20}{2} = 190$. We multiply that by $p^k$ ($0.05^2 = 0.0025$) and $(1-p)^{n-k}$ ($0.95^{18} \approx 0.3972$). 
$190 * 0.0025 * 0.3972 \approx 0.1887$. There is roughly an 18.87% chance of exactly two purchases."

**Common Mistakes Candidates Make:**
* Only calculating $p^k * (1-p)^{n-k}$ and forgetting to multiply by the Binomial Coefficient $\binom{n}{k}$, failing to account for the fact that the 2 purchases could happen at any point among the 20 visitors.

**One Likely Interviewer Follow-up:**
"In Python, how would you modify this to find the probability of getting *at least* 2 purchases?"
*(Answer: I would use the CDF. Specifically, `1 - scipy.stats.binom.cdf(1, 20, 0.05)`, which calculates 1 minus the probability of getting 0 or 1 purchase.)*
