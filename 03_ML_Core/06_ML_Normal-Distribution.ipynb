# Topic: ML | Normal Distribution (Bell Curve)

## Definition (30-second explanation)
* The Normal Distribution (also called Gaussian Distribution or Bell Curve) is the most important probability distribution in statistics and machine learning. 
* It is a continuous, symmetric distribution defined by its mean and standard deviation, where the mean, median, and mode are equal.

## Why Interviewers Ask This
* It underpins fundamental statistical concepts like hypothesis testing, confidence intervals, and the Central Limit Theorem.
* Many ML algorithms, such as linear and logistic regression, assume normally distributed residuals or features.
* It tests your ability to interpret data spread, standardize variables, and apply the empirical rule on the fly.

## Core Concepts
* **Parameters:** Defined entirely by two parameters: the mean ($\mu$) which controls the center, and the standard deviation ($\sigma$) which controls the spread or width.
* **Empirical Rule (68-95-99.7):** Approximately 68% of data falls within $1\sigma$, 95% within $2\sigma$, and 99.7% within $3\sigma$ of the mean.
* **Z-Score:** Calculated as $Z = (X - \mu) / \sigma$, it measures exactly how many standard deviations a data point is from the mean.
* **Central Limit Theorem (CLT):** The distribution of sample means approaches a normal distribution as the sample size grows, regardless of the population's original shape.

## When to Use
* Standardizing features for machine learning models (Z-score normalization).
* Checking assumptions for statistical tests like t-tests or ANOVA.
* Analyzing the distribution of errors (residuals) in regression models.

## Advantages
* Mathematically straightforward and extensively documented across all analytical domains.
* The CLT allows data scientists to leverage normal distribution properties for sample means even when underlying raw data is non-normal.

## Limitations
* The theoretical normal distribution extends from negative infinity to positive infinity, which may not accurately model strictly bounded real-world data.
* Real-world data is frequently skewed, meaning the mean and median can differ significantly, violating normality assumptions.

## Common Comparisons
* **Normal vs. Standard Normal (Z-distribution):** A standard normal distribution is simply a normal distribution that has been standardized to have a mean of 0 ($\mu=0$) and a standard deviation of 1 ($\sigma=1$).

## Common Interview Traps
* **Assuming Normality:** Blindly assuming all data is normally distributed without verifying via Shapiro-Wilk tests, Q-Q plots, or histograms.
* **Applying Parametric Tests to Skewed Data:** Using t-tests or ANOVA on non-normal data without applying a transformation or using non-parametric alternatives.
* **Confusing Mean and Median:** Assuming they are the same in real skewed data just because they are equal in a theoretical normal distribution.

## Example Question and Answer:

### Q1. Employee salaries at a company are normally distributed with mean $\mu = \$75,000$ and standard deviation $\sigma = \$12,000$. Calculate probabilities and Z-scores.

- **(a) What percentage of employees earn between $63,000 and $87,000?**
- **(b) What percentage earn more than $99,000?**
- **(c) What is the z-score of an employee earning $90,000?**

**Ideal Interview Answer:**

* **Part (a):** The range $63,000$ to $87,000$ is exactly $\mu - 1\sigma$ and $\mu + 1\sigma$. By the Empirical Rule, roughly 68.27% of employees fall in this range.

* **Part (b):** $99,000$ is exactly $\mu + 2\sigma$. The Empirical Rule states ~95.45% of data falls within $2\sigma$, leaving 4.55% in the tails. Since the distribution is symmetric, we divide this by 2. Thus, 2.275% of employees earn more than $99,000.

* **Part (c):** The Z-score formula is $(X - \mu) / \sigma$. So, $(90,000 - 75,000) / 12,000 = 1.25$. This salary is 1.25 standard deviations above the mean.

**Common Mistakes Candidates Make:**
* Forgetting to divide the remaining tail probability by 2 in Part (b), accidentally stating that 4.5% of people earn more than $99k (which would include the bottom tail too).
* Not standardizing correctly when computing the Z-score.
