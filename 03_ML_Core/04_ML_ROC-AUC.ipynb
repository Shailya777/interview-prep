# Topic: ROC-AUC (Receiver Operating Characteristic - Area Under Curve)

## Definition (30-second explanation)
* The **ROC curve** is a graph showing a classification model's performance across *all possible classification thresholds*. It plots the True Positive Rate (TPR) against the False Positive Rate (FPR).
* The **AUC (Area Under the Curve)** summarizes this curve into a single score between 0 and 1, representing the model's ability to rank positive instances higher than negative ones.

## Why Interviewers Ask This
* To check if you understand how to evaluate models *before* committing to a specific classification threshold.
* To see if you know the difference between threshold-dependent metrics (Accuracy, F1) and threshold-independent metrics (AUC).
* To test your intuition on when ROC-AUC is inappropriate (e.g., severely imbalanced datasets).

## Core Concepts
* **TPR (True Positive Rate / Recall):** Out of all actual positives, how many did we predict correctly? (TP / TP + FN)
* **FPR (False Positive Rate):** Out of all actual negatives, how many did we falsely predict as positive? (FP / FP + TN)
* **Probabilistic Interpretation:** An AUC of 0.85 means there is an 85% chance that the model will assign a higher predicted probability to a randomly chosen positive instance than to a randomly chosen negative instance.

## When to Use
* When you need a threshold-independent metric to compare the overall ranking power of multiple models.
* When your classes are roughly balanced or moderately imbalanced.
* When you care equally about how well the model distinguishes both the positive and negative classes.

## Advantages
* Separates model performance from threshold selection.
* Robust baseline interpretation: AUC = 0.5 is a random guess; AUC = 1.0 is a perfect classifier.

## Limitations
* **Overly optimistic on highly imbalanced data:** Because FPR uses True Negatives (TN) in its denominator, a massive number of TNs will keep the FPR artificially low, making the ROC curve look falsely excellent.
* Does not give you a direct business cost (unlike calculating dollar loss at a specific threshold).

## Common Comparisons
* **ROC-AUC vs PR-AUC:** ROC-AUC evaluates ranking across both classes; PR-AUC (Precision-Recall AUC) focuses strictly on the minority (positive) class. Use PR-AUC when the positive class is rare (< 5-10%).
* **AUC vs F1-Score:** AUC evaluates the model at *all* thresholds; F1-Score evaluates the model at *one specific* threshold.

## Common Interview Traps
* **Trap 1:** Passing hard class labels (0s and 1s) into the `roc_auc_score` function instead of predicted probabilities (`predict_proba`).
* **Trap 2:** Being happy with an AUC of 1.0. (In the real world, this almost always means data leakage—e.g., including the target variable as a feature).
* **Trap 3:** Using ROC-AUC for a dataset with 0.1% fraud. (Use PR-AUC instead).

## Python / SQL Syntax (if applicable)
* `from sklearn.metrics import roc_auc_score, roc_curve`
* `y_probs = model.predict_proba(X_test)[:, 1]` *(Always use probabilities of the positive class!)*
* `auc_score = roc_auc_score(y_true, y_probs)`
* Finding optimal threshold: `fpr, tpr, thresholds = roc_curve(y_true, y_probs)` -> `optimal_idx = np.argmax(tpr - fpr)`

## Important Formula
* $\text{TPR} = \frac{TP}{TP + FN}$
* $\text{FPR} = \frac{FP}{FP + TN} = 1 - \text{Specificity}$
* $\text{AUC} = P(\text{score}_{pos} > \text{score}_{neg})$

## 45-Second Interview Answer
"ROC-AUC measures a model's ability to rank positive examples higher than negative ones across all possible classification thresholds. The ROC curve plots True Positive Rate versus False Positive Rate, and the AUC is the area underneath it. I use ROC-AUC when I want to compare the general discriminatory power of different models before picking a specific threshold. However, if the dataset is highly imbalanced—like a 1% fraud rate—I switch to Precision-Recall AUC, because ROC-AUC can present an overly optimistic score due to the sheer volume of True Negatives keeping the False Positive Rate low."
