# Topic: ML Regularization (L1/Lasso & L2/Ridge)

## Definition (30-second explanation)
* Regularization is a technique used to prevent overfitting by adding a penalty term to the model's loss function.
* This penalty discourages the model from learning overly complex patterns or memorizing the training data by forcing the model's weights (coefficients) to remain small.

## Why Interviewers Ask This
* To test your fundamental understanding of the bias-variance tradeoff (regularization increases bias to reduce variance).
* To see if you know how to handle high-dimensional data or multicollinearity.
* To ensure you understand the algorithmic differences between feature selection (L1) and coefficient shrinkage (L2).

## Core Concepts
* **Lambda ($\lambda$):** The hyperparameter that controls the strength of the penalty. Higher $\lambda$ = more regularization.
* **L1 Regularization (Lasso):** Penalizes the absolute value of the weights. Can shrink coefficients to exactly zero, effectively performing automatic feature selection.
* **L2 Regularization (Ridge):** Penalizes the squared value of the weights. Shrinks all coefficients smoothly toward zero but rarely exactly to zero.
* **Scikit-Learn's $C$ Parameter:** In sklearn's linear models, $C$ is the inverse of regularization strength ($C = 1/\lambda$). A **smaller** $C$ means **stronger** regularization.

## When to Use
* When your model shows high training accuracy but low test accuracy (overfitting).
* **L1:** When you have many features and suspect a lot of them are irrelevant/redundant (need feature selection).
* **L2:** When all features are known to be somewhat important and you just want to prevent any single feature from dominating the predictions.
* When your dataset is small relative to the number of features.

## Advantages
* Effectively reduces model variance and prevents overfitting.
* L1 yields sparse, highly interpretable models.
* L2 handles multicollinearity well by distributing weights evenly across correlated features.

## Limitations
* Will make an already underfitting model perform even worse (adds unnecessary bias).
* Requires careful hyperparameter tuning for $\lambda$ using Cross-Validation.
* Features **must** be scaled prior to application, otherwise features on larger scales are unfairly penalized.

## Common Comparisons
* **L1 (Lasso) vs L2 (Ridge):** L1 creates sparse models by setting some weights to exactly zero (feature selection). L2 keeps all features but shrinks their weights smoothly (handles correlated features better).

## Common Interview Traps
* **The Scaling Trap:** Forgetting to mention that you must use `StandardScaler` (or similar) before applying regularization. Unscaled features ruin the penalty math.
* **The Sklearn Parameter Trap:** Confusing $C$ with $\lambda$. Remember: smaller $C$ = more penalty.
* **The Guessing Trap:** Stating you would manually pick a $\lambda$ value instead of tuning it systematically via Cross-Validation.

## Example Questions:

### Q1: What is the main difference between L1 (Lasso) and L2 (Ridge) regularization?
* **Ideal Answer:** L1 penalizes the absolute magnitude of the coefficients, which can push some weights to exactly zero, effectively performing automatic feature selection. L2 penalizes the squared magnitude of coefficients, which shrinks them smoothly but rarely to exactly zero, making it better suited for models where all features contribute to the outcome.
* **Common Mistakes:** Mixing up which one performs feature selection. 
* **Follow-up:** Which one would you prefer if your dataset has 10,000 features and you need to explain the top 10 most important ones to stakeholders? (Answer: L1/Lasso, because it will zero out the noise and leave a sparse, interpretable model).

### Q2: Why is feature scaling strictly necessary before applying regularization?
* **Ideal Answer:** Regularization penalizes the magnitude of the model's coefficients. If features are on completely different scales—for example, age in years vs. income in millions—the model will assign artificially small weights to the larger-scale feature just to minimize the loss penalty. Scaling ensures all features are penalized fairly based on their actual predictive importance, not their raw numerical scale.
* **Common Mistakes:** Confusing feature scaling (standardization) with normalization for skewness.
* **Follow-up:** What scaling technique do you typically use? (Answer: StandardScaler, to give all features a mean of 0 and a variance of 1).

### Q3: You are using scikit-learn to train a Logistic Regression model and you notice it's overfitting. You adjust the 'C' parameter from 1.0 to 10.0, but the overfitting gets worse. Why?
* **Ideal Answer:** In scikit-learn, the `C` parameter is the *inverse* of the regularization strength ($\lambda$). By increasing `C` from 1.0 to 10.0, you actually *decreased* the amount of regularization being applied. To fix the overfitting, you need a smaller `C` value, like 0.1 or 0.01, to increase the penalty on the weights.
* **Common Mistakes:** Assuming `C` is exactly the same as $\lambda$ and that higher `C` means more regularization.
* **Follow-up:** How do you find the optimal value for `C` instead of guessing? (Answer: By using a cross-validation strategy, such as `GridSearchCV` or `LogisticRegressionCV`, to test a logarithmic range of C values).
