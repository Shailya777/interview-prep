# Topic: NLP: Word Embeddings (Word2Vec, GloVe, Sentence-BERT)

## Definition (30-second explanation)
* Word Embeddings represent words as dense numerical vectors where semantically similar words are placed close together in vector space.
* Unlike TF-IDF, which creates sparse vectors based on frequency, word embeddings create compact representations capturing relational meaning.

## Why Interviewers Ask This
* To evaluate your understanding of how machines interpret human language semantics beyond simple keyword matching.
* To test your ability to choose the right architecture (static vs. contextual) based on the dataset size and task requirements.

## Core Concepts
* **Static vs. Contextual:** Models like Word2Vec and GloVe assign a single fixed vector per word, whereas BERT generates dynamic vectors based on the surrounding sentence context.
* **Word Analogies:** Embeddings capture spatial relationships (e.g., King - Man + Woman = Queen).
* **Training Architectures:** Word2Vec uses Continuous Bag of Words (CBOW) to predict a target word from context, or Skip-gram to predict context words from a target word.

## When to Use
* **Word2Vec (Skip-gram):** Best when working with rare words.
* **Word2Vec (CBOW):** Best for general word similarity tasks and faster training.
* **Sentence-BERT:** Best for state-of-the-art NLP tasks like semantic search, sentence similarity, and Natural Language Understanding (NLU).

## Advantages
* Dense representation means fixed, manageable dimensionality (typically 50-768 dims) regardless of vocabulary size.
* Captures underlying semantic meaning, not just exact token matches.

## Limitations
* **Static Nature:** Word2Vec assigns the same vector for 'bank' regardless of whether it means a river bank or a financial institution.
* **Out of Vocabulary (OOV):** Standard Word2Vec/GloVe cannot handle words not seen during training; FastText is required to handle OOV via subwords.
* **Interpretability:** Unlike TF-IDF word counts, dense latent spaces have low human interpretability.

## Common Comparisons
* **TF-IDF vs. Word Embeddings:** TF-IDF produces sparse, high-dimensional vectors equal to vocab size without semantic meaning; Embeddings produce dense, fixed-dimension vectors capturing semantics.
* **Word2Vec vs. GloVe:** Word2Vec is predictive (local context windows), while GloVe is count-based (global word co-occurrence matrix).
* **Word2Vec vs. BERT:** Word2Vec is static; BERT is contextual and handles polysemy (multiple meanings of a word).

## Common Interview Traps
* **Mistake 1:** Attempting to train Word2Vec from scratch on tiny datasets instead of using pre-trained models.
* **Mistake 2:** Averaging Word2Vec word embeddings to create sentence embeddings instead of using dedicated Sentence-Transformers.
* **Mistake 3:** Forgetting to normalize embedding vectors before computing cosine similarity.

## Python / SQL Syntax
* **Gensim (Word2Vec):** `model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, sg=1)`.
* **Sentence-Transformers (BERT):** `model = SentenceTransformer('all-MiniLM-L6-v2'); embeddings = model.encode(sentences)`.
* **Similarity Computation:** `cosine_similarity(embeddings[0:1], embeddings[1:2])`.

## 45-Second Interview Answer
"Word embeddings map words into a dense vector space where semantic relationships are preserved as spatial distances. Traditional models like Word2Vec and GloVe generate static embeddings, meaning a word has the exact same representation regardless of its context. To handle words with multiple meanings, modern NLP relies on transformer-based architectures like BERT, which generate contextual embeddings. In practice, I default to pre-trained Sentence-Transformers for sentence similarity tasks, as averaging static word vectors often dilutes the semantic meaning of the sequence."
