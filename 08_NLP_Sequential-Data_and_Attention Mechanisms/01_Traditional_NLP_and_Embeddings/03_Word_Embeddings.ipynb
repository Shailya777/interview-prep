# Topic: NLP: Word Embeddings (Word2Vec, GloVe, Sentence-BERT)

## Definition (30-second explanation)
* Word Embeddings represent words as dense numerical vectors where semantically similar words are placed close together in vector space.
* Unlike TF-IDF, which creates sparse vectors based on frequency, word embeddings create compact representations capturing relational meaning.

## Why Interviewers Ask This
* To evaluate your understanding of how machines interpret human language semantics beyond simple keyword matching.
* To test your ability to choose the right architecture (static vs. contextual) based on the dataset size and task requirements.

## Core Concepts
* **Static vs. Contextual:** Models like Word2Vec and GloVe assign a single fixed vector per word, whereas BERT generates dynamic vectors based on the surrounding sentence context.
* **Word Analogies:** Embeddings capture spatial relationships (e.g., King - Man + Woman = Queen).
* **Training Architectures:** Word2Vec uses Continuous Bag of Words (CBOW) to predict a target word from context, or Skip-gram to predict context words from a target word.

## When to Use
* **Word2Vec (Skip-gram):** Best when working with rare words.
* **Word2Vec (CBOW):** Best for general word similarity tasks and faster training.
* **Sentence-BERT:** Best for state-of-the-art NLP tasks like semantic search, sentence similarity, and Natural Language Understanding (NLU).

## Advantages
* Dense representation means fixed, manageable dimensionality (typically 50-768 dims) regardless of vocabulary size.
* Captures underlying semantic meaning, not just exact token matches.

## Limitations
* **Static Nature:** Word2Vec assigns the same vector for 'bank' regardless of whether it means a river bank or a financial institution.
* **Out of Vocabulary (OOV):** Standard Word2Vec/GloVe cannot handle words not seen during training; FastText is required to handle OOV via subwords.
* **Interpretability:** Unlike TF-IDF word counts, dense latent spaces have low human interpretability.

## Common Comparisons
* **TF-IDF vs. Word Embeddings:** TF-IDF produces sparse, high-dimensional vectors equal to vocab size without semantic meaning; Embeddings produce dense, fixed-dimension vectors capturing semantics.
* **Word2Vec vs. GloVe:** Word2Vec is predictive (local context windows), while GloVe is count-based (global word co-occurrence matrix).
* **Word2Vec vs. BERT:** Word2Vec is static; BERT is contextual and handles polysemy (multiple meanings of a word).

## Common Interview Traps
* **Mistake 1:** Attempting to train Word2Vec from scratch on tiny datasets instead of using pre-trained models.
* **Mistake 2:** Averaging Word2Vec word embeddings to create sentence embeddings instead of using dedicated Sentence-Transformers.
* **Mistake 3:** Forgetting to normalize embedding vectors before computing cosine similarity.

## Python / SQL Syntax
* **Gensim (Word2Vec):** `model = Word2Vec(sentences, vector_size=100, window=5, min_count=1, sg=1)`.
* **Sentence-Transformers (BERT):** `model = SentenceTransformer('all-MiniLM-L6-v2'); embeddings = model.encode(sentences)`.
* **Similarity Computation:** `cosine_similarity(embeddings[0:1], embeddings[1:2])`.

## 45-Second Interview Answer
"Word embeddings map words into a dense vector space where semantic relationships are preserved as spatial distances. Traditional models like Word2Vec and GloVe generate static embeddings, meaning a word has the exact same representation regardless of its context. To handle words with multiple meanings, modern NLP relies on transformer-based architectures like BERT, which generate contextual embeddings. In practice, I default to pre-trained Sentence-Transformers for sentence similarity tasks, as averaging static word vectors often dilutes the semantic meaning of the sequence."

## Practice Questions:

### Q1. What is the difference between Word2Vec CBOW and Skip-gram?
**Answer:** 
CBOW (Continuous Bag of Words) predicts a target word based on its surrounding context words, making it faster and better for frequent words. Skip-gram does the reverse: it uses a single target word to predict the surrounding context words, which makes it slower but much better at capturing representations for rare words.

**Common Mistakes:** 
Confusing which architecture predicts the context vs. which predicts the target word.

**Likely Interviewer Follow-up:** 
If you have a very small dataset with highly specialized medical vocabulary, which one would you choose to train from scratch and why?

**Follow-up Answer:**
Skip-gram, as it consistently produces better representations for rare, highly specialized words compared to CBOW.

### Q2. Why are word embeddings better than one-hot encoding?
**Answer:** 
One-hot encoding creates sparse, orthogonal vectors with a dimensionality equal to the entire vocabulary size, meaning it suffers from the curse of dimensionality and cannot capture semantic relationships (the distance between 'cat' and 'dog' is the same as 'cat' and 'car'). Word embeddings solve this by projecting words into a dense, lower-dimensional space (e.g., 100-300 dimensions) where spatially close vectors represent semantically similar concepts.

**Common Mistakes:** 
Failing to mention the mathematical benefit (dense vs sparse/dimensionality reduction) and focusing only on the "meaning" aspect.

**Likely Interviewer Follow-up:** 
Are there any scenarios where you might still prefer one-hot encoding or TF-IDF over word embeddings?

**Follow-up Answer:**
When exact keyword matching is required, interpretability (knowing exactly which word triggered the score) is critical, or deployment compute resources are extremely constrained.

### Q3. How would you build a sentence similarity system using word embeddings?
**Answer:** 
For modern systems (2024+), I would avoid using static word embeddings (like Word2Vec) and simple averaging, as that loses word order and context. Instead, I would use a pre-trained Sentence-Transformer model (like `all-MiniLM-L6-v2` or a BERT variant) to encode the sentences directly into dense vectors. Then, I would calculate the cosine similarity between those normalized vectors to determine their semantic closeness.

**Common Mistakes:** 
Suggesting to average Word2Vec vectors for sentence embeddings, which is an outdated and flawed approach. Forgetting to mention vector normalization for cosine similarity.

**Likely Interviewer Follow-up:** 
If you need to search across 10 million sentences for the highest similarity, computing cosine similarity sequentially is too slow. How would you scale this architecture?

**Follow-up Answer:**
I would store the pre-computed embeddings in a vector database like ChromaDB or use an Approximate Nearest Neighbor (ANN) index like FAISS to enable scalable, sub-millisecond similarity search.

### Q4. What is the difference between static and contextual word embeddings?
**Answer:** 
Static embeddings (Word2Vec, GloVe, FastText) assign a single, unchanging vector to a word. For example, the word "bank" has the same vector whether it's in "river bank" or "bank account". Contextual embeddings (ELMo, BERT) dynamically generate the vector representation based on the surrounding sentence, allowing the model to accurately capture polysemy (multiple meanings) depending on how the word is used.

**Common Mistakes:** 
Explaining what BERT is without explicitly defining the "static vs dynamic" vector generation mechanism.

**Likely Interviewer Follow-up:** 
If contextual embeddings are superior, why might a team still choose to deploy a FastText or Word2Vec model in production?

**Follow-up Answer:**
They require significantly less memory and computational power, offering ultra-low latency inference for simple tasks where nuanced context isn't strictly necessary.

### Q5. When would you train your own embeddings vs use pretrained ones?
**Answer:** 
I would default to using pre-trained embeddings (like GloVe or standard BERT) for 95% of general NLP tasks to save compute and leverage their massive training corpora. I would only train my own embeddings from scratch if I am working in a highly specialized domain with heavy jargon (like niche legal, medical, or internal corporate telemetry data) where pre-trained models lack the specific vocabulary, and only if I have a sufficiently large corpus (millions of sentences) to train on.

**Common Mistakes:** 
Suggesting training from scratch on small datasets, which leads to poor, overfit vector spaces.

**Likely Interviewer Follow-up:** 
What if you want the benefits of a pre-trained model but have a moderate amount of domain-specific jargon? How do you bridge that gap?

**Follow-up Answer:**
I would perform domain adaptation by continuing the pre-training phase of an existing model (like BERT) on my specialized corpus before fine-tuning it for the specific downstream task.
