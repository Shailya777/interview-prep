# Topic: ML System Design Framework

## Definition (30-second explanation)
*   **Analogy:** Building an ML system is like building a house. If you just show up with a hammer and start nailing wood together (writing model code), the house will collapse. The 5-step framework is the architectural blueprint that ensures you lay the foundation first.
*   The ML System Design Framework is a standardized 5-step methodology (Business Requirements $\rightarrow$ Data $\rightarrow$ Features $\rightarrow$ Model $\rightarrow$ Serving) used to structure ambiguous, open-ended interview questions.

## Why Interviewers Ask This
*   Interviewers want to see if you are a "mature" engineer. Junior candidates instantly jump to "I will use a LoRA-tuned LLM." Senior candidates ask, "What is the latency requirement, and do we even have the data for this?" before choosing a model.

## Core Concepts (The 5-Step Anatomy)
*   **The Bottleneck:** Ambiguity and time limits. In a 45-minute interview, candidates often ramble and run out of time before explaining how the model actually reaches the user.
*   **The Mechanism (The 5 Steps):**
    1.  **Business Requirements (5 mins):** Define the goal, primary metric, latency SLA, and constraints (e.g., "Is this batch or real-time?").
    2.  **Data Pipeline (10 mins):** Identify data sources, storage (SQL/NoSQL/Vector DB), and handle missing data/imbalance.
    3.  **Feature Engineering & Embeddings (10 mins):** Transform raw data into model inputs (e.g., text chunking, categorical encoding, user embeddings).
    4.  **Model Architecture (10 mins):** Always start with a simple baseline (e.g., Logistic Regression / TF-IDF), then propose the advanced architecture (e.g., Two-Tower RecSys or XGBoost).
    5.  **Serving & Monitoring (10 mins):** Define the deployment (REST API vs. Batch), A/B testing strategy, and dashboard metrics (PSI drift).
*   **The Trade-off:** The hardest part is time management. You must forcefully move yourself to the next step even if you want to talk about model architecture for 30 minutes.

## When to Use
*   Use this framework the exact second an interviewer says, "Design a system that does X..." (e.g., "Design Netflix's recommendation engine," or "Design an enterprise RAG system").

## Advantages
*   Provides a safety net. If you freeze, you just look at which of the 5 steps you are on and move to the next one.
*   Shows cross-functional competence (understanding product, data engineering, and MLOps, not just data science).

## Limitations
*   You cannot memorize a specific system (like fraud detection) because the interviewer will intentionally change the constraints (e.g., "What if we don't have user history?"). You must adapt the framework dynamically.

## Common Comparisons
*   **Data Science vs. ML Engineering Design:** Data Scientists usually over-index on Step 3 (Features) and Step 4 (Models). ML Engineers over-index on Step 2 (Data) and Step 5 (Serving). As an Applied GenAI Engineer, you must balance all five.

## Common Interview Traps
*   **The "Jumping to the Model" Trap:** Proposing a complex Deep Learning model in the first 2 minutes without defining the business metric or data sources.
*   **Forgetting the Baseline:** Interviewers will penalize you if you don't offer a dead-simple, cheap baseline model before suggesting an expensive neural network.

## 45-Second Interview Answer
"When approaching an open-ended ML system design, I strictly follow a 5-step framework to ensure I align with business goals before writing any code. I start by defining the Business Requirements and metrics. Second, I map out the Data Pipeline and storage. Third, I define the Feature Engineering or embedding strategy. Fourth, I propose a Model Architecture, always starting with a simple baseline before adding complexity. Finally, I define the Serving and Monitoring strategy to ensure we can track data drift and latency in production."

## Practice Questions:

### Q1: System Design - Step 1 (Requirements)
**Question:** 
The interviewer asks you to design a personalized finance news feed. What are the top 3 clarifying questions you ask to establish the Business Requirements?

**Answer:**
"Before discussing data or models, I need to define the constraints and goals of this system. 
1. **Primary Metric:** How do we define success? Are we optimizing for user engagement (click-through rate/read time) or company revenue (ad clicks)?
2. **Product Constraints:** How is the feed blended? Is it 100% organic user preferences, or do we need to inject sponsored content and trending company-wide news?
3. **Scale & Latency:** What is our expected traffic volume (DAUs), and what is the strict latency SLA when a user opens the app? This will dictate whether we can compute rankings in real-time or if we must use a batch architecture."

**Common Interview Traps:**
* Assuming the goal is "accuracy." In recommendations, success is usually Click-Through Rate (CTR) or Session Length, not standard ML accuracy.
* Forgetting to ask about Scale. Designing a system for 1,000 users is entirely different from designing one for 10 million users.
