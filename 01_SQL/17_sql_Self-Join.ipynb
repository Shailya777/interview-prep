# Topic: SQL Self JOIN (Hierarchical Data & Row Comparison)

## Definition (30-second explanation)
*   A Self JOIN occurs when you join a table to itself by giving it two different aliases. 
*   It is used to compare rows within the same table or to traverse hierarchical data, such as employee-manager relationships.

## Why Interviewers Ask This
*   It requires conceptual thinking about tables as independent instances of data rather than static physical structures.
*   It tests your ability to handle recursive or self-referencing relationships (like `manager_id` pointing to `employee_id`), which are common in real-world database design.

## Core Concepts
*   **Table Aliases:** Mandatory. You must use clear aliases (e.g., `e` for employee, `m` for manager) to distinguish the two instances of the table.
*   **Self-Referencing Keys:** The join condition typically connects a foreign key in the table (like `manager_id`) to the primary key in the same table (like `employee_id`).

## When to Use
*   Querying employee-manager hierarchies.
*   Finding pairs of users who share a common attribute (e.g., same city, same product).
*   Comparing a row's value to other rows in the same table (e.g., days with higher sales than the day before).
*   Detecting symmetric relationships (e.g., User A follows User B, and User B follows User A).

## Advantages
*   Allows you to extract complex relational insights without needing to create and maintain separate physical tables for entities of the same type.
*   Highly efficient for single-level hierarchy traversal (e.g., finding direct reports).

## Limitations
*   Only practical for a fixed number of hierarchical levels (e.g., exactly 1 level up or down). 
*   Can be computationally expensive (resulting in massive cross-joins) if join conditions on large tables are not properly indexed.

## Common Comparisons
*   **Self JOIN vs. Recursive CTE:** A Self JOIN is great for finding an employee's direct manager (1 level deep). A recursive CTE is necessary for finding an employee's entire chain of command up to the CEO (N levels deep).

## Common Interview Traps
*   **Dropping the Root Node:** Using an `INNER JOIN` instead of a `LEFT JOIN` in hierarchy queries. This causes the top-level entity (like a CEO with a `NULL` `manager_id`) to disappear from the results.
*   **Ambiguous Column Names:** Failing to use table aliases when selecting columns, causing SQL to throw an ambiguity error.
*   **Duplicate Pairs:** When finding pairs, failing to use a greater-than/less-than condition (e.g., `e1.id < e2.id`), resulting in mirrored duplicates (counting A-B and B-A as two separate pairs).

## Python / SQL Syntax
```sql
-- Self JOIN to get each employee's manager name
-- LEFT JOIN ensures CEO (NULL manager_id) is still included
SELECT 
    e.employee_id,
    e.name AS employee_name,
    m.name AS manager_name
FROM employees e
LEFT JOIN employees m 
    ON e.manager_id = m.employee_id
ORDER BY e.manager_id NULLS FIRST, e.employee_id;
```

## 45-Second Interview Answer

"A Self JOIN is used when you need to compare rows within the same table or resolve hierarchical relationships, like mapping employees to their managers. The key to writing a Self JOIN is using distinct table aliases so the SQL engine treats the table as two separate entities. In interviews, I always default to using a LEFT JOIN for hierarchies to ensure root nodes—like a CEO who lacks a manager—aren't accidentally filtered out. If I need to traverse an unknown number of levels, I would switch from a Self JOIN to a Recursive CTE."

## Practice Questions:

### Q1: Find Employees Earning More Than Their Manager

**Answer:**
```sql
SELECT 
    e.name AS employee_name,
    e.salary AS employee_salary,
    m.name AS manager_name,
    m.salary AS manager_salary
FROM employees e
JOIN employees m 
    ON e.manager_id = m.employee_id
WHERE e.salary > m.salary;
```

**Common Mistakes Candidates Make:**

**Alias Confusion:** Reversing the salary comparison logic (e.g., WHERE m.salary > e.salary) because they lose track of which alias represents the employee and which represents the manager.

**Using LEFT JOIN Unnecessarily:** While LEFT JOIN is critical for listing all employees, if the strict goal is comparing salaries, an INNER JOIN is fine here because a CEO (with no manager) cannot earn more than their non-existent manager.
