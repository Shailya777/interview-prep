# Topic: PCA Dimensionality Reduction

## Definition (30-second explanation)
* PCA (Principal Component Analysis) is an unsupervised linear transformation technique.
* It compresses high-dimensional data into a smaller set of new features called Principal Components.
* These components are orthogonal and capture the maximum variance in the data.
* It discards the least important information to combat the curse of dimensionality.

## Why Interviewers Ask This
* To test your understanding of the "curse of dimensionality" and how to handle datasets with too many features.
* To verify you know critical preprocessing steps, specifically standardizing data before dimensionality reduction.
* To evaluate if you understand the trade-off between model performance/speed and feature interpretability.

## Core Concepts
* **Standardization:** Features must be scaled (mean=0, std=1) so large-scale features don't dominate the components.
* **Covariance Matrix & Eigenvectors:** PCA computes the covariance matrix and its eigenvectors to find the directions of maximum variance.
* **Eigenvalues:** Represent the amount of variance explained by each principal component.
* **Explained Variance:** The cumulative sum of eigenvalues helps determine how many components to keep (usually targeting 95%+).

## When to Use
* When your dataset has high dimensionality, causing models to train too slowly.
* When features are highly correlated with each other (multicollinearity).
* When you need to visualize high-dimensional data in 2D or 3D.
* When using distance-based algorithms (like KNN, K-Means) that suffer from high dimensions.

## Advantages
* Significantly speeds up algorithm training times and reduces memory footprints.
* Removes multicollinearity by creating orthogonal (uncorrelated) components.
* Filters out noise by dropping components with negligible variance.

## Limitations
* **Loss of Interpretability:** Principal components are linear combinations of original features, making them hard to explain to business stakeholders.
* **Linear Assumption:** Only captures linear relationships; fails on complex non-linear manifolds.
* **Information Loss:** Discarding components inherently means losing some information from the original dataset.

## Common Comparisons
* **PCA vs. Feature Selection:** Feature selection keeps original features; PCA creates entirely new combinations of features.
* **PCA vs. LDA:** PCA is unsupervised (ignores target labels); LDA is supervised and maximizes class separability.
* **PCA vs. t-SNE:** PCA is linear and used for ML inputs; t-SNE is non-linear and used strictly for visualization.
* **PCA vs. Autoencoders:** PCA handles linear compression; Autoencoders capture complex, non-linear patterns.

## Common Interview Traps
* **Mistake 1:** Forgetting to standardize data before applying PCA.
* **Mistake 2:** Applying PCA before the train/test split, which causes data leakage.
* **Mistake 3:** Keeping a fixed number of components instead of using cumulative explained variance to choose $k$.
* **Mistake 4:** Using PCA when the business requires interpretable feature names.
