# Topic: Linear Regression Interpretation

## Definition (30-second explanation)
*   Linear regression models the linear relationship between a continuous dependent variable ($Y$) and one or more independent variables ($X$).
*   It estimates coefficients (betas) that describe the expected change in $Y$ for a one-unit change in $X$, assuming all other variables remain constant.
*   Ordinary Least Squares (OLS) finds these coefficients by minimizing the sum of squared residuals (the difference between predicted and actual values).

## Why Interviewers Ask This
*   To test your foundational statistical literacy and ensure you don't just blindly import `sklearn`.
*   To evaluate your ability to translate mathematical weights into actionable business insights (coefficient interpretation).
*   To check if you understand model validity guardrails, specifically the OLS assumptions, before deploying a model.

## Core Concepts
*   **Coefficients ($\beta_i$):** Represents the change in $Y$ for a 1-unit increase in $X_i$, holding all other features constant. 
*   **R-squared ($R^2$):** The proportion of variance in the target variable that is explained by the model (ranging from 0 to 1).
*   **Adjusted R-squared:** A modified $R^2$ that penalizes the addition of irrelevant features to prevent overfitting.
*   **LINE Assumptions:** The 4 pillars of OLS—Linearity, Independence (of errors), Normality (of residuals), and Homoscedasticity (constant variance of residuals).

## When to Use
*   When predicting a continuous outcome (e.g., house prices, salaries, monthly revenue).
*   As a baseline model before experimenting with more complex, black-box machine learning algorithms.
*   When strict interpretability is required by the business or regulatory bodies (e.g., explaining exactly why a credit limit was set).

## Advantages
*   **Highly Interpretable:** Stakeholders can easily understand the direct impact of each feature via its coefficient.
*   **Computationally Cheap:** Very fast to train and score, even on relatively large datasets.
*   **Statistical Inference:** Provides rich metrics (p-values, confidence intervals) to measure the statistical significance of relationships.

## Limitations
*   **Assumption Heavy:** Fails or provides misleading coefficients if the LINE assumptions are violated.
*   **Sensitive to Outliers:** Large outliers can heavily skew the best-fit line since OLS minimizes *squared* errors.
*   **Multicollinearity:** When features are highly correlated, coefficients become wildly unstable and lose their interpretability.

## Common Comparisons
*   **Linear vs. Logistic Regression:** Linear predicts a continuous target; Logistic predicts a binary/categorical probability.
*   **R-squared vs. Adjusted R-squared:** $R^2$ always increases (or stays the same) as you add features; Adjusted $R^2$ decreases if the new feature doesn't improve the model enough to justify the complexity.
*   **Sklearn vs. Statsmodels:** `sklearn` is optimized for machine learning and predictions; `statsmodels` is optimized for statistical inference and detailed summary reports.

## Common Interview Traps
*   **Interpreting the Intercept literally:** Explaining the intercept as a real-world baseline when $X=0$ is logically impossible (e.g., a house with 0 square feet).
*   **Confusing R-squared with Accuracy:** Stating that an $R^2$ of 0.7 means "70% of predictions are correct" (it actually means 70% of the *variance* is explained).
*   **Ignoring Assumptions:** Trusting a high $R^2$ without plotting residuals to check for heteroscedasticity or non-linearity.

## Important Formula
*   **Linear Equation:** $Y = \beta_0 + \beta_1 X_1 + ... + \beta_p X_p + \epsilon$
*   **R-squared:** $1 - \frac{SS_{res}}{SS_{tot}}$
*   **Adjusted R-squared:** $1 - \frac{(1-R^2)(n-1)}{n-p-1}$ *(where n=samples, p=features)*

## 45-Second Interview Answer
Linear regression models the linear relationship between a continuous target and its features. The most critical outputs are the coefficients, which tell us exactly how much the target changes for a one-unit change in a feature, holding everything else constant. However, for these coefficients to be statistically valid, the model must pass the four LINE assumptions: Linearity, Independence, Normality, and Homoscedasticity. In practice, I evaluate the model using Adjusted R-squared rather than standard R-squared, as it penalizes unnecessary complexity and helps prevent overfitting.
