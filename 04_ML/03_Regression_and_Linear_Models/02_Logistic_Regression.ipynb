# Topic: Logistic Regression Interpretation

## Definition (30-second explanation)
*   Logistic Regression models the probability of a binary outcome (0 or 1, Yes or No).
*   Unlike linear regression, it predicts a probability between 0 and 1 by passing a linear combination of features through a sigmoid function.
*   The raw coefficients represent the change in the log-odds of the target outcome for a one-unit change in the feature.

## Why Interviewers Ask This
*   To test if you can translate complex mathematical outputs (log-odds) into stakeholder-friendly business metrics (odds ratios).
*   To evaluate your critical thinking around classification thresholds (e.g., knowing not to blindly accept the default 0.5 cutoff).
*   To see if you understand how to evaluate models properly when dealing with imbalanced datasets.

## Core Concepts
*   **Log-Odds (Logit):** The linear combination of features. Formula: $\log(P/(1-P)) = \beta_0 + \beta_1 X_1 + \dots + \beta_p X_p$.
*   **Odds Ratio:** Calculated by exponentiating the coefficient ($\exp(\beta_i)$). It represents how the odds of the outcome are multiplied per 1-unit increase in $X_i$.
*   **Threshold:** The probability decision cutoff above which we predict $Y=1$ (default is 0.5).
*   **Sigmoid Function:** The function that maps any real number to the $(0, 1)$ range.

## When to Use
*   For binary classification tasks like churn prediction, fraud detection, or default prediction.
*   When you need probability estimation (e.g., $P(\text{customer buys})$) rather than just a hard class label.
*   As a strong, highly interpretable baseline model before trying complex ensembles like Random Forest or XGBoost.

## Advantages
*   **Interpretability:** Exponentiating the coefficients gives direct multiplicative odds, making it easy to explain feature impact to stakeholders.
*   **Outputs Probabilities:** Provides calibrated probabilities by default, which is crucial for risk thresholding.

## Limitations
*   Assumes a linear relationship between the independent variables and the *log-odds* of the outcome.
*   Cannot natively handle non-linear decision boundaries without manual feature engineering.
*   Can overfit if the sample size is very small relative to the number of features.

## Common Comparisons
*   **Logistic vs Linear Regression:** Logistic predicts probabilities bounded between 0 and 1 using a sigmoid curve; Linear predicts continuous, unbounded values.
*   **predict() vs predict_proba():** `predict()` outputs the hard class label (0 or 1) based on a threshold, while `predict_proba()` outputs the actual calculated probability.

## Common Interview Traps
*   **Interpreting Coefficients as Probabilities:** A common fatal mistake is stating that a coefficient of 0.5 means a 50% increase in probability. It is a change in *log-odds*.
*   **Defaulting to 0.5 Threshold:** Using the default 0.5 without optimizing for the specific business cost of False Positives vs. False Negatives.
*   **Using Accuracy on Imbalanced Data:** Citing accuracy as the main metric for imbalanced datasets instead of ROC-AUC, F1-score, or PR-AUC.

## 45-Second Interview Answer
Logistic regression models the probability of a binary event occurring, like customer churn, by applying a sigmoid function to a linear combination of features. The most important thing to remember is that the raw model coefficients represent changes in log-odds, not probabilities. To make these actionable for business stakeholders, we exponentiate the coefficients to get Odds Ratios, which tell us exactly how a 1-unit change in a feature multiplies the odds of the outcome. Additionally, I never rely solely on the default 0.5 threshold or standard accuracy; I tune the threshold based on the business cost of errors and evaluate using ROC-AUC or PR-AUC.
