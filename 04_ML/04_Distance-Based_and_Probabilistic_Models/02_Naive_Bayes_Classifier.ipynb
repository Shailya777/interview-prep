# Topic: ML: Naive Bayes Classifier

## Definition (30-second explanation)
The Naive Bayes Classifier is a fast, probabilistic machine learning algorithm based on Bayes' Theorem. It is called "naive" because it assumes that all features are conditionally independent of each other given the class label—a strong assumption that is rarely true in real life but surprisingly effective for high-dimensional classification tasks like text analysis.

## Why Interviewers Ask This
*   **Probability Fundamentals:** It tests your foundational knowledge of conditional probability and Bayes' Theorem.
*   **Zero-Frequency Problem:** It allows interviewers to ask about Laplace (additive) smoothing.
*   **Generative vs. Discriminative:** It is the classic example of a Generative model, usually contrasted against Logistic Regression (Discriminative).
*   **NLP Baselines:** It proves you know how to build a quick, standard baseline for text classification (e.g., spam detection) before throwing heavy LLMs at a problem.

## Core Concepts
*   **Bayes' Theorem:** Calculates the posterior probability $P(Y|X)$ using the prior probability $P(Y)$, the likelihood $P(X|Y)$, and the evidence $P(X)$.
*   **The Naive Assumption:** Assumes features $x_1, x_2, ... x_n$ are independent given $Y$. This changes the joint likelihood $P(x_1, x_2...|Y)$ into a simple product: $\prod P(x_i|Y)$.
*   **Laplace Smoothing:** A technique to handle unseen features during inference. If a word wasn't in the training data, its probability becomes 0, zeroing out the entire product. Laplace smoothing adds a small value (usually $\alpha=1$) to the numerator and the vocabulary size to the denominator.
*   **Log Probabilities:** Because multiplying many tiny probabilities causes numerical underflow, Naive Bayes implementations sum the logarithms of probabilities instead.

## When to Use
*   As a first-pass baseline model for text classification (Spam, Sentiment Analysis, Topic Categorization).
*   When training data is relatively small but feature dimensionality is very high.
*   When inference speed and low compute resources are top priorities.

## Advantages
*   Extremely fast to train and predict (linear time complexity $O(N \times D)$).
*   Highly scalable with high-dimensional data (e.g., TF-IDF vectors with 100k+ features).
*   Requires very little training data to estimate the necessary probabilities.
*   Robust to irrelevant features.

## Limitations
*   **The Independence Assumption:** It ignores feature correlations (e.g., the words "San" and "Francisco" are highly correlated, but NB treats them as independent).
*   **Bad Estimator:** While good at *ranking* classes to make a prediction, the actual probability outputs (e.g., "99% sure this is spam") are notoriously unreliable and poorly calibrated.
*   Continuous variables require assumptions about their distribution (usually Gaussian), which may not fit the real data.

## Common Comparisons
*   **Naive Bayes vs. Logistic Regression:** NB is a Generative model (models the joint distribution $P(X,Y)$); Logistic Regression is a Discriminative model (models the decision boundary $P(Y|X)$ directly). LR generally performs better when data is large and features are correlated, while NB converges faster with less data.
*   **Multinomial vs. Gaussian vs. Bernoulli NB:** 
    *   *Multinomial:* For discrete counts (e.g., word frequencies/TF-IDF).
    *   *Bernoulli:* For binary/boolean features (e.g., word presence/absence).
    *   *Gaussian:* For continuous features (assumes features follow a normal distribution).

## Common Interview Traps
*   **Forgetting Laplace Smoothing:** If asked what happens when a new word appears in the test set, failing to mention the probability dropping to zero and the Laplace smoothing fix is an automatic red flag.
*   **Using the wrong variant:** Proposing Gaussian NB for sparse text TF-IDF arrays (should be Multinomial).
*   **Not mentioning log-probabilities:** Failing to mention log-addition to solve floating-point underflow when multiplying thousands of small probabilities.

## Important Formula (if applicable)
*   **Bayes' Theorem:** $$P(Y|X) = \frac{P(X|Y)P(Y)}{P(X)}$$
*   **Naive Bayes Classifier Rule:** $$\hat{y} = \arg\max_y P(y) \prod_{i=1}^n P(x_i | y)$$
*   **Laplace Smoothing (for word $w$ in class $c$):** $$P(w|c) = \frac{\text{count}(w, c) + \alpha}{\text{count}(c) + \alpha |V|}$$ *(where $|V|$ is the vocabulary size)*
