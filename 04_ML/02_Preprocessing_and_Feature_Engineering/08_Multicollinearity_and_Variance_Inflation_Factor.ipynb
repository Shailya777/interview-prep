# Topic: Multicollinearity & Variance Inflation Factor (VIF)

## Definition (30-second explanation)
* Multicollinearity occurs when two or more independent variables (features) in a regression model are highly correlated with each other.
* This prevents the model from isolating the individual effect of each correlated variable, making the coefficients unstable and unreliable.

## Why Interviewers Ask This
* To test your foundational understanding of linear regression assumptions and diagnostics.
* To verify if you know how to handle feature selection, model interpretability, and unstable models in a real-world business context.

## Core Concepts
* **Variance Inflation Factor (VIF):** Quantifies how much the variance of a coefficient is inflated due to multicollinearity.
* **Thresholds:** VIF = 1 means perfect (no multicollinearity). VIF > 5 is moderate (investigate). VIF > 10 indicates severe multicollinearity (take action).
* **Symptoms:** Inflated standard errors, misleading/unreliable p-values (e.g., high R-squared but all p-values > 0.05), and unexpected coefficient signs.

## When to Use (Detection)
* Before fitting any regression model with multiple features.
* When you observe large coefficient changes upon adding or removing features.
* During feature engineering, especially when creating derived features from existing ones.

## Handling Multicollinearity
* **Drop Features:** Remove one of the highly collinear features based on business importance.
* **Regularization:** Use Ridge Regression (L2 penalty) because it inherently handles multicollinearity by shrinking correlated coefficients together.
* **Dimensionality Reduction / Feature Engineering:** Use PCA to combine correlated features into uncorrelated principal components, or create a composite feature (e.g., size/rooms).

## Limitations
* **Prediction vs. Interpretation:** Multicollinearity does NOT hurt prediction accuracy; it only hurts coefficient interpretation.
* **Model Immunity:** Tree-based models (like Random Forest and XGBoost) do not use linear assumptions and are not affected by multicollinearity in the same way.

## Common Comparisons
* **OLS vs. Ridge:** Ordinary Least Squares (OLS) struggles with multicollinearity, whereas Ridge Regression is robust to it due to the L2 penalty.

## Common Interview Traps
* **The Pairwise Trap:** Checking only pairwise correlations; multicollinearity can exist among 3+ features even if pairwise correlation ($r$) is low.
* **The Blind Drop:** Dropping features based solely on VIF > 10 without considering their business importance or predictive power.
* **The Prediction Myth:** Believing that multicollinearity degrades the model's overall predictive performance.

## Important Formula
$$VIF_i = \frac{1}{1 - R_i^2}$$
*(Where $R_i^2$ is the R-squared value obtained when regressing feature $X_i$ against all other independent features)*

## 45-Second Interview Answer
"Multicollinearity happens when features in a regression model are highly correlated, causing standard errors to inflate and making the coefficients highly unstable. While it doesn’t hurt the model's overall prediction accuracy, it ruins our ability to interpret feature importance. I detect it using the Variance Inflation Factor (VIF); a VIF over 10 indicates a severe issue. To resolve it, I typically drop the least important collinear feature, combine them using PCA, or switch to an L2 regularized model like Ridge regression which handles collinearity naturally."
