# Topic: Decision Tree | Impurity Criteria & Overfitting Controls

## Definition (30-second explanation)
A Decision Tree is a supervised machine learning algorithm that makes decisions by splitting data into subsets based on feature values. It acts like a flowchart, finding splits that maximize information gain to create pure leaf nodes.

## Why Interviewers Ask This
* Tests your understanding of tree-based algorithms (the foundation for Random Forests and XGBoost).
* Evaluates your grasp of the bias-variance tradeoff (specifically how to handle overfitting).
* Assesses your ability to explain models to non-technical stakeholders (interpretability).

## Core Concepts
* **Node Purity:** The goal of splitting is to group similar instances together. A "pure" node contains only one class.
* **Impurity Measures:** Metrics (Gini or Entropy) used to decide which feature to split on at each node.
* **Information Gain:** The reduction in impurity achieved by splitting the data.
* **Hyperparameter Pruning:** Techniques to restrict tree growth (e.g., controlling depth or node size) to prevent learning noise.

## When to Use
* When you need a highly interpretable, explainable model for stakeholders.
* When data has non-linear relationships that linear models fail to capture.
* When you want automatic feature selection out of the box.
* When dealing with mixed data types (categorical and numerical) with minimal preprocessing.

## Advantages
* **Interpretable:** The decision-making process can be easily visualized.
* **Low Preprocessing:** Does NOT require feature scaling (normalization/standardization).
* **Feature Importance:** Automatically highlights the most useful features for splitting.

## Limitations
* **Prone to Overfitting:** An unconstrained tree will perfectly memorize the training data, capturing noise.
* **Instability:** Small variations in the training data can result in a completely different tree structure.

## Common Comparisons
* **Gini vs. Entropy:** Gini is computationally faster (no log function) and is the default in scikit-learn. Entropy can sometimes yield slightly more balanced/accurate trees, but the practical difference in performance is usually negligible.
* **Decision Tree vs. Random Forest:** A single tree is a weak learner prone to high variance (overfitting); Random Forest ensembles multiple trees to reduce variance and improve generalization.

## Python / SQL Syntax

    from sklearn.tree import DecisionTreeClassifier
    
    # Initialize model with overfitting controls
    dt = DecisionTreeClassifier(
        criterion="gini",
        max_depth=4,              # Limits tree depth
        min_samples_split=5,      # Min samples required to split an internal node
        min_samples_leaf=2,       # Min samples required to be at a leaf node
        random_state=42
    )
    dt.fit(X_train, y_train)

## Important Formula
* **Gini Impurity:** $Gini = 1 - \sum (p_i^2)$ (where $p_i$ is the proportion of class $i$)
* **Entropy:** $Entropy = -\sum (p_i \cdot \log_2(p_i))$
* **Information Gain:** Information Gain = Impurity(parent) - Weighted Average Impurity(children)

## 45-Second Interview Answer
"A Decision Tree is a non-linear algorithm that recursively splits data to maximize Information Gain, creating pure subsets. At each node, it evaluates features using an impurity metric like Gini or Entropy. While they are highly interpretable and require no feature scaling, single trees are notoriously prone to overfitting. In practice, we mitigate this by tuning structural hyperparameters like `max_depth` and `min_samples_split`, or by graduating to ensemble methods like Random Forests for better generalization."
