# Topic: Conditional Probability

## Definition (30-second explanation)
Conditional probability measures the likelihood of an event occurring given that another specific event has already occurred. It allows us to update our probability estimates based on new information, restricting our sample space to only the cases where the condition is true.

## Why Interviewers Ask This
*   It is the absolute mathematical foundation for Naive Bayes and Bayesian networks.
*   It is required to truly understand ML evaluation metrics (e.g., Precision is the conditional probability of a true positive *given* a positive prediction).
*   It tests your ability to avoid common cognitive biases, such as the base rate fallacy.

## Core Concepts
*   **The Formula:** $P(A \vert{} B) = P(A \text{ and } B) / P(B)$.
*   **Independence Test:** Two events are independent if knowing one gives no new information about the other. Mathematically: $P(A \vert{} B) = P(A)$.
*   **Restricted Sample Space:** When computing $P(A\vert{}B)$, your denominator changes from the "entire universe" to just the "universe where B happened."

## When to Use
*   When extracting insights from contingency tables (cross-tabulations).
*   When calculating the probability of a sequence of dependent events.
*   When updating beliefs or model weights based on new arriving data.

## Advantages
*   Provides a rigorous framework for dealing with partial information.
*   Forms the basis of Bayes' Theorem, unlocking advanced probabilistic machine learning.

## Limitations
*   Heavily relies on having an accurate measure of the base condition $P(B)$. 
*   If $P(B) = 0$, the conditional probability is undefined.

## Common Comparisons
*   **Conditional vs. Joint Probability:** Joint $P(A \text{ and } B)$ is the probability both happen in the whole population. Conditional $P(A \vert{} B)$ is the probability A happens *within the sub-population* of B.
*   **$P(A \vert{} B)$ vs. $P(B \vert{} A)$:** They are rarely equal. The probability of having an umbrella given it is raining is high; the probability of it raining given you have an umbrella is much lower.

## Common Interview Traps
*   **Assuming $P(A \vert{} B) = P(B \vert{} A)$:** This is a classic trap (e.g., confusing the probability of having a disease given a positive test, with a positive test given the disease).
*   **Confusing it with Joint Probability:** Using the total dataset size as the denominator instead of just the subset where condition B is true.
