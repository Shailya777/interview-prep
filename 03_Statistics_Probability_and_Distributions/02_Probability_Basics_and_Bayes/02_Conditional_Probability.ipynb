# Topic: Conditional Probability

## Definition (30-second explanation)
Conditional probability measures the likelihood of an event occurring given that another specific event has already occurred. It allows us to update our probability estimates based on new information, restricting our sample space to only the cases where the condition is true.

## Why Interviewers Ask This
*   It is the absolute mathematical foundation for Naive Bayes and Bayesian networks.
*   It is required to truly understand ML evaluation metrics (e.g., Precision is the conditional probability of a true positive *given* a positive prediction).
*   It tests your ability to avoid common cognitive biases, such as the base rate fallacy.

## Core Concepts
*   **The Formula:** $P(A \vert{} B) = P(A \text{ and } B) / P(B)$.
*   **Independence Test:** Two events are independent if knowing one gives no new information about the other. Mathematically: $P(A \vert{} B) = P(A)$.
*   **Restricted Sample Space:** When computing $P(A\vert{}B)$, your denominator changes from the "entire universe" to just the "universe where B happened."

## When to Use
*   When extracting insights from contingency tables (cross-tabulations).
*   When calculating the probability of a sequence of dependent events.
*   When updating beliefs or model weights based on new arriving data.

## Advantages
*   Provides a rigorous framework for dealing with partial information.
*   Forms the basis of Bayes' Theorem, unlocking advanced probabilistic machine learning.

## Limitations
*   Heavily relies on having an accurate measure of the base condition $P(B)$. 
*   If $P(B) = 0$, the conditional probability is undefined.

## Common Comparisons
*   **Conditional vs. Joint Probability:** Joint $P(A \text{ and } B)$ is the probability both happen in the whole population. Conditional $P(A \vert{} B)$ is the probability A happens *within the sub-population* of B.
*   **$P(A \vert{} B)$ vs. $P(B \vert{} A)$:** They are rarely equal. The probability of having an umbrella given it is raining is high; the probability of it raining given you have an umbrella is much lower.

## Common Interview Traps
*   **Assuming $P(A \vert{} B) = P(B \vert{} A)$:** This is a classic trap (e.g., confusing the probability of having a disease given a positive test, with a positive test given the disease).
*   **Confusing it with Joint Probability:** Using the total dataset size as the denominator instead of just the subset where condition B is true.

## 45-Second Interview Answer
"Conditional probability is how we update the likelihood of an event when we receive new evidence. Instead of looking at the entire sample space, we restrict our view to only the scenarios where our condition is true. The formula is the joint probability of both events divided by the probability of the condition. It's incredibly important in machine learning because it is the driving logic behind Bayesian models and helps us correctly interpret classification metrics."

## Example Questions:

### Q1:

**Question:**
In a dataset of 1000 users: 400 are male, 600 are female. 300 of the 400 males purchased a product, and 200 of the 600 females purchased. What is P(purchased | male)?

**Ideal Interview Answer:**
We are looking for the probability of a purchase *given* the user is male. We restrict our denominator to only the males.
$P(\text{purchased} \vert{} \text{male}) = \frac{P(\text{purchased and male})}{P(\text{male})} = \frac{300/1000}{400/1000} = \frac{300}{400} = 0.75$, or 75%.


**Common mistakes candidates make:**
*   Dividing 300 by the total number of users (1000) instead of the total number of males, incorrectly calculating the joint probability $P(\text{purchased and male}) = 0.30$.

**One likely interviewer follow-up:**
*   "Based on this data, are 'gender' and 'purchasing' independent events? Prove it mathematically."

### Q2

**Question:**
A disease affects 1% of the population. A test for this disease is 90% accurate for positive cases and 95% accurate for negative cases. If a person tests positive, what is the probability they actually have the disease?

**Ideal Interview Answer:**
This requires translating conditional probabilities into Bayes Theorem. 
We need $P(\text{Disease} | \text{Positive})$.
We know:

$P(\text{Disease}) = 0.01$

$P(\text{No Disease}) = 0.99$

$P(\text{Positive} | \text{Disease}) = 0.90$ (True Positive Rate)

$P(\text{Positive} | \text{No Disease}) = 0.05$ (False Positive Rate, which is $1 - 0.95$)

$P(\text{Disease} | \text{Positive}) = \frac{P(\text{Positive} | \text{Disease}) \times P(\text{Disease})}{P(\text{Positive})}$

$P(\text{Positive}) = (0.90 \times 0.01) + (0.05 \times 0.99) = 0.009 + 0.0495 = 0.0585$

Result: $0.009 / 0.0585 = 15.4\%$.

**Common mistakes candidates make:**
*   Falling for the base rate fallacy: Assuming the answer is 90% because the test is "90% accurate," completely ignoring the massive pool of healthy people generating false positives.

**One likely interviewer follow-up:**
*   "If we wanted to increase the final probability to 50%, what would we need to change about the test's False Positive Rate?"

## Practice Questions:

### Q1:
You are analyzing data for an upcoming marketing push. A product manager looks at your dataset and claims, "Knowing a user is on a mobile device changes their probability of converting. They are dependent! We need a mobile-specific ad campaign."

**Question:**
Write the Pandas code to calculate the probabilities necessary to prove or disprove the product manager's claim. What specific mathematical condition must be true for you to tell the PM they are wrong (i.e., that device and conversion are actually independent)?

**Answer:**
"To prove the events are independent, we must mathematically show that $P(\text{converted} | \text{mobile}) = P(\text{converted})$. In other words, the conversion rate for mobile users must be exactly the same as the overall conversion rate. 

In Pandas, we can check this by comparing the mean of the entire converted column to the mean of the filtered subset:

````python
# Data:
import pandas as pd
data = {'device': ['mobile', 'desktop', 'mobile', 'desktop', 'mobile'], 
        'converted': [1, 0, 0, 1, 1]}
df = pd.DataFrame(data)
# Assume the real df has 100,000 rows.
````

````python
# 1. Overall Probability: P(converted)
overall_conversion_rate = df['converted'].mean()

# 2. Conditional Probability: P(converted | mobile)
mobile_users = df[df['device'] == 'mobile']
mobile_conversion_rate = mobile_users['converted'].mean()

# 3. Independence Check
is_independent = (overall_conversion_rate == mobile_conversion_rate)
````
