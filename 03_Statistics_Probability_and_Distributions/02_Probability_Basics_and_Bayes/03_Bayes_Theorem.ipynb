# Topic: Bayes Theorem & Base Rate Fallacy

## Definition (30-second explanation)
* Bayes' Theorem is a mathematical formula that calculates how to update the probability of a hypothesis (our belief) when given new evidence or data.
* It connects prior knowledge with observed data to produce an updated, posterior probability.

## Why Interviewers Ask This
* It is the foundational concept behind probabilistic machine learning models, A/B testing, and risk assessment.
* Interviewers want to test your statistical intuition and see if you fall for the "Base Rate Fallacy" (ignoring the underlying prevalence of an event).

## Core Concepts
* **Prior ( P(A) ):** Your initial belief about the probability of an event before seeing any new evidence.
* **Likelihood ( P(B|A) ):** The probability of observing the new evidence if your initial hypothesis is true.
* **Marginal/Evidence ( P(B) ):** The total probability of observing the evidence under all possible scenarios (often calculated using the Law of Total Probability).
* **Posterior ( P(A|B) ):** Your updated belief about the hypothesis after factoring in the new evidence.

## When to Use
* When assessing the true reliability of diagnostic tests, anomaly detection systems, or classification models.
* When establishing baselines for Naive Bayes classifiers.

## Advantages
* Provides a rigorous mathematical framework for updating beliefs continuously as new data streams in.
* Forces practitioners to explicitly define and account for their prior assumptions.

## Limitations
* Requires an accurate prior probability, which can sometimes be subjective or difficult to estimate accurately in the real world.
* Naive Bayes (which builds on this) assumes features are completely independent, which is rarely true in practice.

## Common Comparisons
* **Bayesian vs. Frequentist:** Frequentists treat probability as the long-run frequency of repeatable events (no priors). Bayesians treat probability as a degree of belief that updates with evidence.

## Common Interview Traps
* **The Base Rate Fallacy:** Assuming that a highly accurate test (e.g., 95% accuracy) guarantees a high probability of a true positive when the condition is extremely rare (e.g., 1% prevalence). 
* **Confusing Likelihood and Posterior:** Mixing up P(Test Positive | Disease) with P(Disease | Test Positive).

## Python / SQL Syntax
```python
# Calculating Posterior using Bayes' Theorem
prior_disease = 0.01
prior_no_disease = 0.99
sensitivity = 0.95  # P(Pos | Disease)
false_positive_rate = 0.10  # P(Pos | No Disease)

# Law of Total Probability
p_evidence = (sensitivity * prior_disease) + (false_positive_rate * prior_no_disease)

# Bayes Theorem
posterior = (sensitivity * prior_disease) / p_evidence
print(f"Posterior Probability: {posterior:.4f}") 
```

## Important Formula
* **Bayes Theorem:** $P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$
* **Law of Total Probability:** $P(B) = P(B|A) \cdot P(A) + P(B|\text{not } A) \cdot P(\text{not } A)$

## 45-Second Interview Answer
"Bayes' Theorem provides a mathematical way to update our beliefs based on new evidence. It calculates the posterior probability by multiplying the prior probability by the likelihood of the evidence, divided by the total probability of that evidence occurring. It is critical in data science because it prevents the Base Rate Fallacy—reminding us that if a condition is inherently rare, even a highly accurate model will produce a large number of false positives."

## Practice Questions:

### Q1: Spam Filter Probability
**Question:** Your spam filter flags 5% of emails as spam. Of the truly spam emails, the filter catches 92%. Of legitimate emails, it incorrectly flags 3%. If an email is flagged, what is the probability it is actually spam?

**Answer:**
Mathematically, we define our variables:
* P(Spam) = 0.05
* P(Flag | Spam) = 0.92
* P(Flag | Legit) = 0.03
* P(Legit) = 0.95

Using Law of Total Probability: P(Flag) = (0.92 * 0.05) + (0.03 * 0.95) = 0.046 + 0.0285 = 0.0745
Using Bayes: P(Spam | Flag) = (0.92 * 0.05) / 0.0745 = 0.617 (or 61.7%)

**Common Mistakes Candidates Make:**
* Ignoring the base rate (5%) and simply assuming the probability is 92% because the test is highly sensitive.
* Forgetting to multiply the false positive rate (3%) by the probability of the negative class (95%) when calculating the denominator.

**Likely Interviewer Follow-up:** 
"If we wanted to increase this posterior probability to 80% without changing the underlying algorithm's sensitivity or specificity, what would have to happen in the real world?" *(Answer: The base rate of spam emails entering the system would have to increase).*
