# Topic: Power Analysis & Minimum Detectable Effect (MDE)

## Definition (30-second explanation)
* **Statistical Power** is the probability of correctly rejecting a false null hypothesis—essentially, the chance of detecting a real effect when one exists. 
* Power = $1 - \beta$ (where $\beta$ is the Type II error rate). 
* **Power Analysis** determines the minimum sample size ($n$) required to detect a specific Minimum Detectable Effect (MDE) given a set power level and significance level ($\alpha$).

## Why Interviewers Ask This
* To ensure you know how to design statistically sound experiments before launching them.
* To test if you understand the business trade-off: detecting smaller effects requires exponentially more time, users, and money.
* To see if you can correctly interpret a non-significant A/B test result (checking if it was underpowered).

## Core Concepts
* **Alpha ($\alpha$)**: Significance level or Type I error rate (industry standard is usually 0.05).
* **Power ($1 - \beta$)**: Probability of detecting a real effect (industry standard is usually 0.80).
* **MDE ($\delta$)**: Minimum Detectable Effect; the smallest difference between control and treatment that is practically significant for the business.
* **Variance ($\sigma^2$)**: The spread of the outcome variable; higher variance requires a larger sample size.

## When to Use
* **Before** running any A/B test or clinical trial to determine the required sample size.
* When planning experiment timelines (calculating how many days a test needs to run based on daily traffic).
* When investigating a non-significant result to check if the test actually had the sensitivity to detect the expected effect.

## Advantages
* Prevents running underpowered studies that waste resources and yield inconclusive results.
* Saves money by avoiding over-collecting data beyond what is statistically necessary.
* Forces product and business teams to align on what constitutes a "meaningful" impact (MDE) beforehand.

## Limitations
* Requires estimating the baseline variance and defining the MDE *before* data collection.
* **Post-hoc power analysis** (calculating power using the observed effect size after the test) is mathematically circular and misleading.

## Common Comparisons
* **Cohen's h vs. Cohen's d**: Use Cohen's h for standardized effect sizes of *proportions* (e.g., conversion rates). Use Cohen's d for *continuous* metrics (e.g., revenue per user).
* **Power vs. Confidence Level**: Power ($1 - \beta$) is about avoiding false negatives. Confidence Level ($1 - \alpha$) is about avoiding false positives.

## Common Interview Traps
* **Setting the MDE too small**: Interviewers will test if you know that a tiny MDE (e.g., 0.1%) requires millions of users. 
* **Saying a test "failed" just because $p > 0.05$**: Always state that you must check if the test was adequately powered first.
* **Calculating post-hoc power**: Never recommend doing this using the observed effect size.

## Python Syntax 

    from statsmodels.stats.power import NormalIndPower
    from statsmodels.stats.proportion import proportion_effectsize
    import numpy as np

    # 1. Calculate effect size (Cohen's h)
    effect_size = proportion_effectsize(p2_expected, p1_baseline)

    # 2. Solve for sample size
    analysis = NormalIndPower()
    n_per_group = analysis.solve_power(effect_size=effect_size, alpha=0.05, power=0.80, alternative='two-sided')

    # 3. Always round up
    total_users_needed = int(np.ceil(n_per_group)) * 2


## Important Formula 
$n = \frac{(z_{\alpha/2} + z_{\beta})^2 \cdot 2 \cdot \sigma^2}{\delta^2}$
*(Where $z_{\alpha/2}$ is 1.96 for $\alpha=0.05$, and $z_{\beta}$ is 0.842 for Power=0.80)*

## 45-Second Interview Answer
"Power analysis is a prerequisite step in A/B testing used to determine the minimum sample size required for an experiment. It relies on four interlocked variables: the significance level, the statistical power, the data's variance, and the Minimum Detectable Effect. By fixing alpha at 0.05, power at 0.80, and estimating our baseline variance, we can calculate exactly how many users we need to detect the MDE. This ensures we don't waste resources on underpowered tests or run tests longer than necessary."

## Practice Questions:

### Q1:
**Question:** Your A/B test ran for 7 days with 500 users per group and showed p=0.08. What might have gone wrong?

**Answer:**
The test was likely underpowered. A sample size of 500 users per group is very small for most digital A/B tests unless the expected effect size is massive. Because $p=0.08$ is close to significance, the test might have detected a real effect if it had more power (larger $n$). A non-significant result with low power is inconclusive. We should have conducted a power analysis before starting to determine the correct runtime and sample size.

**Common Mistakes Candidates Make:**
* Stating definitively that "the treatment had no effect" or "the null hypothesis is true."
* Suggesting we just "run the test for a few more days" (peeking/p-hacking) without a proper power analysis plan.

**Interviewer Follow-up:**
"If you can't run the test any longer due to business constraints, how else could you increase the statistical power?" (Answer: You can increase power by accepting a larger Minimum Detectable Effect (MDE), increasing your significance level ($\alpha$), or reducing metric variance using techniques like CUPED.)

### Q2:
**Question:** If you double the sample size, what happens to statistical power?

**Answer:**
Increasing the sample size decreases the standard error of the estimate. Because the distributions of the null and alternative hypotheses become narrower (less variance), there is less overlap between them. This increases the statistical power, meaning we have a higher probability of correctly rejecting the null hypothesis if a true effect exists.

**Common Mistakes Candidates Make:**
* Saying power exactly "doubles" (power follows a non-linear curve and asymptotes at 1.0).
* Confusing power with the alpha level.

**Interviewer Follow-up:**
"According to the sample size formula, how does sample size scale with the Minimum Detectable Effect (delta)?" (Answer: Sample size scales inversely with the square of the MDE ($1/\delta^2$), meaning halving your target effect size requires four times the sample size.)

### Q3:

**Question:** What is Cohen's d? When would you use it vs Cohen's h?

**Answer:**
Both are standardized measures of effect size used in power analysis. Cohen's d is used for comparing the means of two continuous variables (e.g., Average Order Value, time spent on page). It is calculated as the difference between means divided by the pooled standard deviation. Cohen's h, on the other hand, is used for comparing two proportions (e.g., Click-Through Rate, Conversion Rate) and relies on the arcsine transformation of those proportions.

**Common Mistakes Candidates Make:**
* Using Cohen's d for conversion rates (proportions).
* Forgetting that effect size must be standardized relative to variance/standard deviation before calculating power.

**Interviewer Follow-up:**
"If you are using `statsmodels` in Python to calculate required sample size for a revenue metric, which power class would you instantiate?" (Answer: You would instantiate the TTestIndPower class because revenue is a continuous metric, whereas conversion rates require NormalIndPower for proportions.)

### Q4:
**Question:** Name four factors that increase the statistical power of a test.

**Answer:**
1. **Increasing the Sample Size ($n$)**: Reduces standard error, narrowing the distributions.
2. **Increasing the Effect Size / MDE ($\delta$)**: Pushes the alternative distribution further away from the null.
3. **Decreasing the Variance ($\sigma^2$)**: Narrower distributions mean less overlap.
4. **Increasing the Significance Level ($\alpha$)**: Moving $\alpha$ from 0.05 to 0.10 makes it easier to reject the null, which mathematically increases power (though it increases false positives).

**Common Mistakes Candidates Make:**
* Failing to mention variance/standard deviation.
* Incorrectly stating that *decreasing* alpha increases power (they are inversely related).

**Interviewer Follow-up:**
"Out of those four, which one is usually out of your control as an experimenter?" (Answer: The variance ($\sigma^2$) of the underlying user behavior is inherently out of your direct control.)

### Q5: Sample Size & Runtime Calculation
**Scenario:** Calculate total users and days required for an A/B test given a baseline CR of 12%, expected CR of 14%, 2,000 daily traffic, $\alpha=0.05$, and power=0.80.

```python
import math
from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize
    
baseline_cr = 0.12
expected_cr = 0.14
daily_traffic = 2000
    
# 1. Calculate standardized effect size (Cohen's h)
effect_size = proportion_effectsize(expected_cr, baseline_cr)
    
# 2. Solve for sample size (returns n per group!)
analysis = NormalIndPower()
n_per_group = analysis.solve_power(effect_size=effect_size, 
                                       alpha=0.05, 
                                       power=0.80, 
                                       alternative='two-sided')
    
# 3. Calculate total users needed (always round UP)
total_users = math.ceil(n_per_group) * 2
    
# 4. Calculate full days required
days_required = math.ceil(total_users / daily_traffic)
    
print(f"Total Users: {total_users}")
print(f"Days Required: {days_required}")
```

**Interview Tips & Common Traps:**
* **TRAP 1:** Using absolute difference (0.14 - 0.12) as the effect size. Always use `proportion_effectsize` for proportions.
* **TRAP 2:** Forgetting to multiply by 2 (or number of variants) to get total traffic.
* **TRAP 3:** Using `round()` instead of `ceil()`. You must always round up to guarantee minimum power.

### Q6: 
Scenario: Negotiating A/B Test Timelines

**Question:** Your test requires 6 weeks, but the PM demands it be done in 3. How do you handle this and what parameters can you negotiate?

**Ideal Interview Answer:**
"I cannot magically reduce the timeline without changing the statistical risk profile or the target impact. If we must cut the runtime in half, I would present the PM with three specific trade-offs:
1. **Increase the MDE (Target a larger effect):** We can run it in 3 weeks, but we will only have the power to detect a much larger lift (e.g., 4% instead of 2%). If the actual lift is only 2%, we will likely miss it.
2. **Accept higher False Positive Risk (Increase Alpha):** We can increase $\alpha$ from 0.05 to 0.10 or 0.20. We will reach a conclusion faster, but we significantly increase the risk of launching a feature that actually does nothing.
3. **Accept higher False Negative Risk (Decrease Power):** We can drop power from 80% to 60%. This means if the feature truly works, we have a 40% chance of failing to detect it and throwing away a good idea.
Ultimately, I would recommend increasing the MDE. We should ask ourselves: 'If the lift is less than X%, do we even care to launch it?'"

**Common Mistakes Candidates Make:**
* Mixing up the consequences of Alpha and Power (e.g., saying higher alpha causes false negatives).
* Saying "just add more users" without acknowledging that daily traffic is usually a fixed physical constraint.
