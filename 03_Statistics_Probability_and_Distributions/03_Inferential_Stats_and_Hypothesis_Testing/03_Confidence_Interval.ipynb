# Topic: Confidence Interval (CI)

## Definition (30-second explanation)
* A Confidence Interval provides a range of plausible values for an unknown population parameter, based on sample data.
* A 95% CI means: if we were to repeat this sampling procedure 100 times, we expect about 95 of those calculated intervals to contain the true population parameter. 

## Why Interviewers Ask This
* It is the most commonly misinterpreted concept in statistics. Interviewers want to see if you fall for the "95% probability" trap.
* It tests your understanding of the relationship between sample size, variance, and the precision of your estimates.

## Core Concepts
* **Margin of Error (ME):** The amount added and subtracted from the sample mean to create the CI. $ME = z^* \times SE$.
* **Z* Critical Values:** The multiplier that determines the confidence level (e.g., 1.96 for 95% confidence, 1.645 for 90%, 2.576 for 99%).
* **CI Width Drivers:** 
    * *Larger sample size ($n$)* = Narrower CI (More precision)
    * *Lower variance ($\sigma$)* = Narrower CI 
    * *Higher confidence (e.g., 99% vs 95%)* = Wider CI (More certainty requires a wider net)

## When to Use
* To estimate a metric (like average session time or conversion rate) when you only have a sample of data.
* To determine statistical significance in A/B testing (if the CI for the *difference* between two groups does not contain 0, the result is statistically significant).

## Advantages
* Provides a measure of uncertainty (a range) rather than just a misleading single point estimate.
* Directly actionable for business stakeholders (e.g., "The true uplift is between +1% and +3%").

## Limitations
* Highly sensitive to outliers, which can inflate the standard deviation and make the interval impractically wide.
* Assumes the sample is truly random and representative of the population (no selection bias).

## Common Comparisons
* **Confidence Interval vs. Point Estimate:** A point estimate is a single guess (e.g., sample mean = 5). A CI is a range (e.g., [4.1, 5.9]) that quantifies the uncertainty of that guess.
* **Z-distribution vs. T-distribution:** Use $z^*$ when the population standard deviation ($\sigma$) is known or $n \ge 30$. Use $t^*$ when $\sigma$ is unknown and $n$ is small (requires degrees of freedom, $n-1$).

## Common Interview Traps
* **The "Probability" Trap:** Never say "There is a 95% probability that the true mean is in this specific interval." The true mean is a fixed number—it is either in the interval (100%) or it isn't (0%). The 95% refers to the *process* of generating intervals.

## Important Formula (if applicable)
* **Confidence Interval:**
  $$CI = \bar{x} \pm z^* \left( \frac{\sigma}{\sqrt{n}} \right)$$
  *(Substitute $z^*$ with $t^*$ and $\sigma$ with sample std dev $s$ if population variance is unknown)*

## 45-Second Interview Answer
"A Confidence Interval gives us a range of plausible values for a population parameter based on our sample. It's calculated by taking the sample mean and adding/subtracting a margin of error. The biggest trap is saying 'there's a 95% chance the true mean is in here.' Instead, 95% confidence means if we repeated the experiment 100 times, 95 of those intervals would capture the true parameter. In A/B testing, if the 95% CI for the difference between a control and treatment group doesn't cross zero, we can declare the result statistically significant at an alpha of 0.05."

## Practice Questions:

### Q1:
**Question:** 
"In an A/B test, the control group has a mean conversion rate of 5% and the treatment group has 5.8%. The 95% CI for the difference is [0.3%, 1.5%]. Can you conclude the treatment is better? How do you use the CI here?"

**Answer:**
Yes, we can conclude the treatment is statistically significantly better. We use the Confidence Interval here by checking if it contains zero. Since the entire 95% CI [0.3%, 1.5%] is strictly above zero, it means "no effect" (a difference of 0) is not a plausible value. The treatment increases conversion by somewhere between 0.3% and 1.5% with 95% confidence. This is mathematically equivalent to rejecting the null hypothesis at an alpha level of 0.05.

**Common Mistakes Candidates Make:**
* Stating "The treatment is better because 5.8% > 5%" without referencing the Confidence Interval's role in proving it wasn't just random chance.
* Saying "There is a 95% chance the uplift is between 0.3% and 1.5%."

**One Likely Interviewer Follow-up:**
* "What if the 95% CI for the difference was [-0.1%, 1.2%]? What would you tell the Product Manager?"
*(Answer: I would tell them the result is not statistically significant because the interval contains zero, meaning the treatment might actually be worse or have no effect. We cannot confidently roll out the feature based on this test.)*
