# Topic: Null vs Alternative Hypothesis

## Definition (30-second explanation)
* The Null Hypothesis ($H_0$) is the default assumption that there is no effect, no difference, or no relationship. It represents the status quo.
* The Alternative Hypothesis ($H_1$ or $H_a$) is the claim you are actively trying to prove—representing a real effect or difference.

## Why Interviewers Ask This
* It forms the foundational backbone of all inferential statistics and A/B testing.
* Interviewers use it to test your scientific rigor: Do you understand how to properly frame a research question before looking at the data?

## Core Concepts
* **Equality Constraint:** $H_0$ must always be formulated as a specific equality (e.g., parameter = claimed_value). 
* **One-Tailed vs. Two-Tailed:** A two-tailed test checks for differences in *both* directions ($H_1: \mu \neq \mu_0$). A one-tailed test checks a specific direction ($H_1: \mu > \mu_0$ or $H_1: \mu < \mu_0$).
* **Alpha ($\alpha$):** The significance level (commonly 0.05) representing the threshold to reject $H_0$. This must be defined *before* running the test.
* **The Decision Rule:** If the p-value is less than alpha, you reject $H_0$.

## When to Use
* Testing if a new feature improves a business metric (A/B testing).
* Checking if a new ML model performs significantly better than a random baseline.
* Comparing two distinct groups to see if they differ significantly.

## Advantages
* Provides a strict, objective, and mathematical framework for validating business assumptions with data.
* Prevents teams from making costly changes based on random noise in the data.

## Limitations
* Hypothesis testing requires a specific, well-defined question before data collection.
* It is not suitable for exploratory data analysis or when the sample size is too small to detect meaningful effects.

## Common Comparisons
* **Failing to Reject $H_0$ vs. Proving $H_0$:** You can never *prove* the Null Hypothesis is true; you can only conclude there isn't enough evidence to reject it.

## Common Interview Traps
* **Setting $H_0$ as a range:** Stating "$H_0: \mu > 30$" is incorrect; $H_0$ must be a specific baseline equality like "$H_0: \mu = 30$".
* **Confusing $H_0$ and $H_1$:** Always set $H_0$ as the default (no effect) and $H_1$ as your new claim.
* **Setting $\alpha$ after looking at the data:** This is a classic form of p-hacking. Alpha must be set beforehand.

## Python / SQL Syntax (if applicable)
* **One-sample t-test using SciPy:**
  ```python
  from scipy import stats
  # deliveries = array of sample data, claimed_mean = H0 value
  t_stat, p_value = stats.ttest_1samp(deliveries, claimed_mean) #
  ```

## 45-Second Interview Answer
"In any statistical test, the Null Hypothesis ($H_0$) is our default assumption of 'no effect' or 'status quo', formulated as an exact equality. The Alternative Hypothesis ($H_1$) is the actual effect or difference we are trying to prove. We run our test to see if we have enough statistical evidence to reject $H_0$. The biggest rule to remember is that we can never *prove* $H_0$ is true—we only either reject it or fail to reject it based on our p-value and pre-defined alpha."

## Practice Questions:

### Q1:

**Question 1:**
Explain the difference between H0 and H1. How do you decide which is which?

**Answer:**
$H_0$ is the null hypothesis—the default assumption of no effect or no difference. $H_1$ is the alternative hypothesis—the effect or difference you are trying to prove. You always set $H_0$ to represent the status quo (e.g., "The new model is the same as the old model") and $H_1$ to represent your research claim (e.g., "The new model is better"). 

**Common Mistakes Candidates Make:**
* Swapping them, which fundamentally breaks the math of the test since the test assumes $H_0$ is true when calculating the p-value.

**One Likely Interviewer Follow-up:**
* "What happens if your p-value is 0.06 and your alpha is 0.05?" *(Answer: We fail to reject $H_0$. We do not have sufficient evidence to support our claim ($H_1$))*

### Q2:
**Question 2:**
A company claims average call wait time is 5 minutes. You sample 40 calls, mean=6.2 min, std=1.5. State H0 and H1. What test do you use?

**Answer:**
* $H_0: \mu = 5$ minutes
* $H_1: \mu \neq 5$ minutes (assuming a two-tailed test to check if it's different in either direction)
* Since the population standard deviation is unknown, we would typically use a one-sample t-test. However, because $n \ge 30$, a z-test would also be a mathematically acceptable approximation.

**Common Mistakes Candidates Make:**
* Stating $H_0: \mu = 6.2$. ($H_0$ uses the *claimed* population parameter, not the sample statistic).

**One Likely Interviewer Follow-up:**
* "If you wanted to specifically prove the wait times are *worse* (longer) than 5 minutes, how would $H_1$ change?" *(Answer: It would become a one-tailed right-sided test: $H_1: \mu > 5$)*

### Q3:
**Question 3:**
What is wrong with this statement: H0: The new feature increases revenue?

**Answer:**
This statement places the *claim* in the Null Hypothesis. $H_0$ must always represent the status quo of "no effect." The correct formulation is $H_0$: The new feature has *no effect* on revenue, and $H_1$: The new feature increases revenue. 

**Common Mistakes Candidates Make:**
* Failing to recognize that $H_0$ requires an equality constraint mathematically.

**One Likely Interviewer Follow-up:**
* "Why do we force $H_0$ to be 'no effect'?" *(Answer: It gives us a fixed mathematical distribution centered at zero difference, allowing us to calculate the probability (p-value) of observing our sample data.)*

### Q4:
**Question 4:**
Can you ever PROVE that H0 is true? Explain your answer.

**Answer:**
No, you can never prove $H_0$ is true. In hypothesis testing, we either "reject $H_0$" or "fail to reject $H_0$". Failing to reject $H_0$ just means our current sample didn't provide enough evidence to overturn the status quo; it doesn't definitively prove the status quo is an absolute truth.

**Common Mistakes Candidates Make:**
* Concluding a test by saying "Therefore, we accept the null hypothesis."

**One Likely Interviewer Follow-up:**
* "If you fail to reject $H_0$, does that mean your experiment was a failure?" *(Answer: Not necessarily. Knowing a feature or model does *not* have an impact is still valuable business information that saves deployment costs.)*

### Q5:
**Question:**
Your Product Manager suggests using a one-tailed test because it will reach statistical significance faster, requiring less data, and they only care if the metric goes up. Do you agree? What are the statistical and business risks of exclusively using a one-tailed test here?

**Answer:**
"I would advise against it and recommend a two-tailed test. While the PM is correct that a one-tailed test lowers the critical value and requires less data to detect a positive shift, it introduces a major business risk: it is completely blind to negative impacts. If the new feature contains a bug and performs *significantly worse*, a one-tailed test will simply output 'fail to reject the null'—lumping a neutral result and a catastrophic failure into the exact same bucket. To protect our metrics, we need a two-tailed test to detect significant movements in *either* direction."

**Interview Tips:**
* **Industry Standard:** Always default to two-tailed tests in interviews unless explicitly asked otherwise. It is the gold standard at companies like Meta, Amazon, and Netflix because the cost of shipping a broken feature outweighs the speed gained by a one-tailed test.
* **Alpha Allocation:** Mentioning that a two-tailed test splits the alpha (e.g., 2.5% on each side for $\alpha=0.05$) shows deep mechanical understanding.

### Q6:
**Question:**
You are evaluating a new Generative AI model's generation speed. The current production model has a known average latency of 850 milliseconds. You generate 100 responses using the new model (stored in `new_model_latencies`). Write Python code to run a hypothesis test to see if the new model's latency is significantly different, extract the p-value, and make a deployment decision based on $\alpha = 0.05$.

````python
import numpy as np
from scipy import stats

np.random.seed(42)
new_model_latencies = np.random.normal(loc=830, scale=90, size=100)
baseline_latency = 850
````

````python
import numpy as np
from scipy import stats
# OR: from statsmodels.stats.weightstats import ztest

baseline_latency = 850

# Method 1: Using SciPy's 1-sample t-test
t_stat, p_value = stats.ttest_1samp(new_model_latencies, baseline_latency)

# Check for significance AND that the latency actually improved (decreased)
if p_value < 0.05 and np.mean(new_model_latencies) < baseline_latency:
    print("Deploy new model")
else:
    print("Keep current model")
````


**Interview Tips:**
* **The "Two-Sided" Trap:** When automating decisions based on two-sided tests, always check the direction of the sample mean! A p-value < 0.05 just means "different." If you don't check if the new metric is actually *better*, you might automate the deployment of a significantly *worse* model.
* **T-test vs Z-test:** For large sample sizes ($n \ge 30$), the t-distribution and z-distribution converge. Using `statsmodels` `ztest` or `scipy` `ttest_1samp` will yield nearly identical decisions. SciPy is just slightly more common in quick coding interviews.
