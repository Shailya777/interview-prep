# Topic: p-value Interpretation

## Definition (30-second explanation)
* The p-value is the probability of observing a test statistic as extreme as, or more extreme than, the one computed from your sample data, assuming the null hypothesis ($H_0$) is true.
* It is a measure of how surprising the data would be if $H_0$ were actually true.
* It is explicitly NOT the probability that $H_0$ is true, nor is it the probability that your result is correct.

## Why Interviewers Ask This
* It is widely considered one of the most misunderstood concepts in statistics.
* Interviewers need to know you will not launch a product feature based on a fundamental misinterpretation of test results.
* To ensure you understand the difference between an effect being "real" (statistical significance) and an effect being "large" (practical significance).

## Core Concepts
* **Decision Rule:** If p-value < alpha, you reject $H_0$. If p-value >= alpha, you fail to reject $H_0$.
* **Alpha ($\alpha$):** The significance level or threshold, commonly set at 0.05 or 0.01, used to make the rejection decision.
* **Statistical vs Practical Significance:** A result can have a tiny p-value (e.g., < 0.001) in a massive sample size, but the actual effect size might be practically irrelevant to the business.
* **Effect Size Context:** Always pair p-values with an effect size metric (like Cohen's d) to provide a complete picture.

## When to Use
* After running statistical tests (t-test, z-test, chi-square, ANOVA) to evaluate the null hypothesis.
* In A/B testing to determine if the observed difference is real or just due to chance.
* In regression analysis or feature selection to evaluate if a coefficient or variable has a statistically significant relationship with the target.

## Limitations & When NOT to Use
* Do not rely solely on p-values without reporting the effect size.
* Be highly cautious when sample sizes are extremely large, as microscopic, meaningless effects will become statistically significant.
* Never use a p-value to measure the magnitude (size) of an effect.
* Do not use uncorrected p-values across multiple tests (use Bonferroni or FDR corrections to avoid Type I error inflation).

## Common Comparisons
* **p-value vs. Effect Size:** The p-value tells you if an effect exists; the effect size tells you how large and meaningful it is.
* **p-value vs. Alpha:** The p-value is computed from your specific sample data. Alpha is the pre-defined risk threshold you set before running the test.

## Common Interview Traps
* **TRAP:** Stating that a p-value of 0.05 means there is a 5% chance the null hypothesis is true. 
* **TRAP:** Saying "the p-value is 0.03, so there is a 97% chance our result is correct.".
* **TRAP:** Testing multiple segments/features and reporting the single significant p-value without applying a Bonferroni or Benjamini-Hochberg (BH) correction.

## Important Formula 
$p\text{-value} = P(\text{observing test stat} \ge \text{observed} \mid H_0 \text{ is true})$

## 45-Second Interview Answer
"The p-value is the probability of observing data at least as extreme as ours, assuming the null hypothesis is completely true. It is a measure of surprise: a small p-value, typically under our alpha of 0.05, means our data is highly unlikely under the null, giving us evidence to reject it. However, a p-value only indicates statistical significance, not practical significance. Therefore, I always pair the p-value with an effect size to communicate both the reality and the magnitude of the business impact."

## Practice Questions:

### Q1:
**Question:** What does p=0.08 mean if alpha=0.05? What is your decision?

**Ideal Interview Answer:**
A p-value of 0.08 means that if the null hypothesis were true, there is an 8% chance of observing a result at least this extreme. Because 0.08 is greater than our pre-determined alpha threshold of 0.05, we fail to reject the null hypothesis. The result is not statistically significant.

**Common Mistakes Candidates Make:**
* Saying "we accept the null hypothesis." (You can only *fail to reject* it).
* Arguing that it is "almost significant" and we should just run the test a little longer to push it under 0.05 (this is p-hacking).

**Interviewer Follow-up:**
"If you strongly suspect the effect is real despite the p=0.08, what might have been wrong with your experimental design?" (Answer: The test was likely underpowered due to an insufficient sample size or a much higher-than-expected variance in the baseline metric.)
