# Topic: Hypothesis Testing

## Definition (30-second explanation)
* Hypothesis testing is the process of using data to test a statistical claim about a population. 
* It forms the statistical backbone of data-driven decisions in business, including A/B testing and clinical trials.
* The framework relies on establishing a null hypothesis (H0), which represents the default or status quo, and an alternative hypothesis (H1), which is the effect you want to prove.
* You calculate a p-value to determine the probability of observing your data assuming H0 is true, and if p < alpha, you reject H0.

## Why Interviewers Ask This
* To ensure you follow a rigorous, objective framework (the 7-step process) rather than making gut-based decisions.
* To test your ability to correctly select the right statistical test based on data types and assumptions (e.g., t-test vs. chi-square).
* To see if you understand the critical business difference between statistical significance and practical (business) significance.

## Core Concepts
* **Alpha & p-value:** Alpha is the maximum false positive rate tolerated (usually 0.05), while the p-value is the probability of seeing data this extreme if H0 is true.
* **Type I vs Type II Errors:** Type I error (alpha) is a false positive (rejecting a true H0), and Type II error (beta) is a false negative (failing to reject H0 when H1 is true).
* **Statistical vs Practical Significance:** Statistical significance means p-value < alpha, whereas practical significance means the effect size is actually large enough to matter to the business.
* **Power (1-beta):** The probability of correctly detecting a real effect.

## When to Use
* Use a **Z-test or t-test** to compare one mean to a known value, depending on sample size.
* Use a **Two-sample t-test (Welch)** to compare means of two independent groups (e.g., continuous metrics in A/B testing).
* Use a **Z-test for proportions** when comparing binary outcomes (e.g., conversion rates) between two groups.
* Use a **Chi-square test** to test independence of categorical variables using frequency data.

## Advantages
* Provides a standardized, mathematically sound 7-step framework for decision-making.
* Prevents teams from launching features based on random noise by establishing strict alpha thresholds.

## Limitations
* Highly sensitive to sample size; a tiny, practically meaningless effect can become statistically significant with a large enough 'n'.
* Relies heavily on underlying assumptions (normality, independence, equal variance) that must be checked.

## Common Comparisons
* **Z-test vs T-test:** Z-tests are for proportions or means with known sigma/large n; t-tests are for means with unknown sigma.
* **One-tailed vs Two-tailed:** Two-tailed tests look for any difference (positive or negative), while one-tailed looks only in a specific direction (using a one-tailed test when two-tailed is appropriate is a common mistake).
* **ANOVA vs t-test:** Use ANOVA for comparing means across 3 or more groups, and a t-test for comparing exactly 2 groups.

## Common Interview Traps
* **"Accepting" the null hypothesis:** You must never accept H0; the correct terminology is "fail to reject H0" because absence of evidence is not evidence of absence.
* **P-hacking:** Defining your hypotheses (H1) *after* seeing the data, rather than before.
* **Multiple Testing Problem:** Running multiple tests and using alpha=0.05 for each without applying a correction (like Bonferroni), which drastically inflates your Type I error rate.

## Python Syntax 

    import numpy as np
    from scipy import stats
    from statsmodels.stats.proportion import proportions_ztest
    
    # 1. Two-sample Welch t-test (Continuous data, unequal variance)
    #
    t_stat, p_val = stats.ttest_ind(control_group, treatment_group, equal_var=False) 
    
    # 2. Z-test for proportions (Binary data)
    #
    counts = np.array([conversions_control, conversions_treatment])
    nobs = np.array([total_control, total_treatment])
    z_stat, p_val_prop = proportions_ztest(counts, nobs)

## Important Formula 
* **Z-test (Known Sigma):** $Z = \frac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$
* **One-Sample t-test (Unknown Sigma):** $t = \frac{\bar{x} - \mu_0}{s / \sqrt{n}}$ with $df = n - 1$

## 45-Second Interview Answer
"Hypothesis testing is a 7-step statistical framework used to make data-driven decisions, heavily utilized in A/B testing. It involves setting a default null hypothesis (H0) and an alternative hypothesis (H1) before collecting data to avoid p-hacking. By calculating a p-value using the appropriate test—like a Welch t-test for continuous means or a Z-test for proportions—we evaluate the probability of observing our data assuming H0 is true. If the p-value falls below our pre-defined alpha threshold, usually 0.05, we reject H0, though we must always verify if the statistical result carries practical, business significance."

## Practice Questions:

### Q1:
**Question:** In an A/B test, treatment group shows p-value = 0.049. Your manager says to ship it immediately. What additional factors should you consider before making a final decision?

**Answer:**
While a p-value of 0.049 is technically statistically significant at the standard alpha=0.05, it is right on the boundary, so I would advise caution. Before shipping, I would check the effect size to ensure practical significance—is the lift actually meaningful to the business? I'd verify the test had adequate statistical power (80%+) and check if we tested multiple metrics or segments, which might require a multiple-testing correction. Finally, I'd consider novelty effects and weigh any business risks or engineering costs associated with deploying the feature.

**Common Mistakes Candidates Make:**
* Blindly agreeing with the manager just because $p < 0.05$.
* Forgetting to mention the distinction between statistical and practical significance.

**Interviewer Follow-up:**
"If you discover that the team tested 10 different metrics during this experiment and this was the only one that showed a p-value < 0.05, how would that change your recommendation?" (Answer: I would advise against shipping because testing 10 metrics drastically inflates the false positive rate, requiring a multiple-testing correction like Bonferroni (which lowers the alpha threshold to 0.005) and would likely render this result statistically insignificant.)
