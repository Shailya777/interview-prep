# Topic: Type I and Type II Error

## Definition (30-second explanation)
* A Type I Error (False Positive) occurs when you reject the null hypothesis even though it is actually true. 
* A Type II Error (False Negative) occurs when you fail to reject the null hypothesis even though it is actually false. 

## Why Interviewers Ask This
* It is essential for designing any A/B test, choosing sample sizes, and understanding model evaluation.
* Interviewers want to see if you can translate statistical errors into real-world business trade-offs.

## Core Concepts
* **Alpha ($\alpha$):** The probability of making a Type I Error (False Positive).
* **Beta ($\beta$):** The probability of making a Type II Error (False Negative).
* **Power ($1 - \beta$):** The probability of correctly detecting a real effect (True Positive).
* **The Trade-off:** Reducing alpha naturally increases beta, and vice versa.

## When to Use
* When discussing trade-offs between false positives and false negatives in classification models.
* When explaining why precision and recall matter in machine learning.
* During the design phase of medical testing, fraud detection, and quality control systems.

## Advantages
* Provides a structured framework to apply cost-benefit analysis to business decisions based on data.
* Allows you to compute exactly how much data is needed to reliably detect an effect by setting target alpha and power levels.

## Limitations
* You can never completely eliminate both errors simultaneously without an infinite sample size; there is always a trade-off.
* Real-world costs of these errors can be difficult to quantify mathematically.

## Common Comparisons
* **Medical Analogy:** Type I = telling a healthy person they are sick. Type II = missing a disease in a sick person.
* **Machine Learning Analogy:** Type I connects directly to False Positives (impacting Precision); Type II connects directly to False Negatives (impacting Recall).

## Common Interview Traps
* **Confusing the errors:** Mixing up which error corresponds to False Positives vs. False Negatives.
* **Setting alpha arbitrarily low:** Candidates often suggest setting alpha to 0.001 to avoid false alarms, forgetting that this drastically increases Type II errors (missing real effects).
* **Forgetting the Power formula:** Failing to remember that Power equals $1 - \beta$.

## Important Formula
- Statistical Power:
$$Power = 1 - \beta$$

- Probability of Type I Error:
$$P(\text{Type I Error}) = \alpha$$

## 45-Second Interview Answer
"In hypothesis testing, a Type I error is a False Positive—rejecting a true null hypothesis, which is controlled by our alpha level. A Type II error is a False Negative—failing to reject a false null hypothesis, controlled by beta. Because reducing one error inherently increases the other, we must look at the business context to decide which is more costly. To reduce both errors simultaneously and increase the overall statistical power of our test, our only option is to increase the sample size."

## Practice Questions:

### Q1:
**Question 1: In a fraud detection model, which error is more costly: Type I or Type II? How does this affect your model design?**

**Answer:**
In fraud detection, a Type II Error (missing a fraudulent transaction — false negative) is typically more costly because the fraud causes immediate, real financial damage. However, too many Type I Errors (false positives — blocking legitimate transactions) can severely hurt customer experience. The trade-off is controlled by adjusting the classification threshold. A lower threshold reduces Type II errors but increases Type I errors. We use the business context to decide the optimal threshold using a cost-benefit analysis.

**Common Mistakes Candidates Make:**
* Focusing exclusively on the financial loss of Type II errors while entirely ignoring the customer churn and support costs associated with Type I errors.

**One Likely Interviewer Follow-up:**
* "Technically, how would you adjust this threshold in a scikit-learn model?" *(Answer: Instead of using `.predict()`, which defaults to a 0.5 threshold, use `.predict_proba()` and apply a custom threshold like 0.2 to prioritize catching fraud.)*

### Q2:
**Question 2: A spam filter incorrectly marks 3% of legitimate emails as spam. What type of error is this?**

**Ideal Interview Answer:**
This is a Type I Error, or a False Positive. The null hypothesis here is that the email is legitimate. The filter incorrectly rejected the null hypothesis and flagged it as spam.

**Common Mistakes Candidates Make:**
* Confusing it with a Type II error (which would be letting actual spam into the inbox).

**One Likely Interviewer Follow-up:**
* "If we want to reduce this 3% error rate, what happens to our Type II error rate?" *(Answer: It will increase. By making the filter stricter to prevent legitimate emails from going to spam, more actual spam will slip through into the inbox.)*

### Q3:
**Question 3: How does increasing alpha from 0.05 to 0.10 affect Type I and Type II error rates?**

**Ideal Interview Answer:**
Increasing alpha directly increases the acceptable Type I error rate (False Positives) from 5% to 10%. Because of the inverse relationship between the two, this will decrease the Type II error rate (False Negatives), making it easier to detect a real effect and thereby increasing the statistical power of the test.

**Common Mistakes Candidates Make:**
* Stating that both errors increase, failing to recognize the trade-off.

**One Likely Interviewer Follow-up:**
* "In what business scenario would you intentionally choose an alpha of 0.10 instead of 0.05?" *(Answer: When missing a true effect (Type II error) is much more expensive than a false alarm, such as early-stage disease screening or exploratory feature testing.)*

### Q4:
**Question 4: You need to detect a small effect reliably. What do you increase: sample size or alpha?**

**Ideal Interview Answer:**
You should increase the sample size. Increasing the sample size gives you more precision, which reduces both Type I and Type II errors simultaneously and increases the power of the test. Increasing alpha would technically give you more power, but at the unacceptable cost of significantly more false positives.

**Common Mistakes Candidates Make:**
* Suggesting increasing alpha as a viable primary strategy for detecting small effects.

**One Likely Interviewer Follow-up:**
* "What if it's impossible to increase the sample size due to traffic constraints?" *(Answer: You must accept lower statistical power, run the test for a longer duration, or target a larger Minimum Detectable Effect (MDE).)*

### Q5:
**Question:**
You are designing an A/B test for a new checkout flow. You want to control your Type I error rate at 5% ($\alpha = 0.05$) and want an 80% chance of correctly detecting the effect if it exists (Power = 0.80). The expected effect size is 0.02. Using `statsmodels`, calculate the minimum sample size required per group.

````python
import math
from statsmodels.stats.power import TTestIndPower

alpha = 0.05
power = 0.80
effect_size = 0.02

# Initialize the power analysis object
analysis = TTestIndPower()

# solve_power automatically solves for whichever parameter is omitted (in this case, nobs1)
required_n = analysis.solve_power(effect_size=effect_size, alpha=alpha, power=power, alternative='two-sided')

# Always round up for sample sizes (you cannot have a fraction of a user)
print(f"Required sample size per group: {math.ceil(required_n)}")
````

**Interview Tips:**
* **The "Four Pillars" of Power Analysis:** Interviewers look to see if you know that Alpha, Power, Effect Size, and Sample Size are mathematically locked together. If you know three, you can always solve for the fourth.
* **Always Round Up:** In experiment design, always use a ceiling function for sample size calculations to ensure you hit your minimum power requirement.
