# Topic: Correlation vs Causation

## Definition (30-second explanation)
* Correlation measures the statistical relationship between two variables, meaning when one changes, the other tends to change in a predictable direction. 
* Causation dictates that a change in one variable directly produces a change in another. 
* Crucially, correlation does not imply causation, as two variables can move together purely by coincidence or due to a hidden third variable.

## Why Interviewers Ask This
* This is considered the most critical distinction in Data Science.
* Interviewers want to ensure you won't mislead stakeholders into making wrong business decisions based on spurious correlations.
* It tests your foundational knowledge for designing experiments (A/B tests) and selecting appropriate features for predictive models.

## Core Concepts
* **Confounder:** A hidden third variable that influences and causes changes in both of the observed variables, creating a false correlation between them.
* **Spurious Correlation:** A situation where two variables appear correlated either by sheer coincidence or because of a confounder.
* **Reverse Causality:** Occurs when an analyst assumes variable X causes variable Y, but in reality, Y causes X (e.g., assuming higher ad spend causes higher revenue, when higher revenue actually funds higher ad budgets).
* **Pearson r Limitation:** Pearson r only measures *linear* correlation and can be 0 even if a strong nonlinear relationship exists.

## When to Use (Establishing Causation)
* **Randomized Controlled Trials (RCTs) / A/B Tests:** This is the gold standard for proving causation.
* **Causal Inference Techniques:** When experiments are impossible, use methods like Difference-in-Differences (DiD), Instrumental Variables (IV), or Regression Discontinuity Design (RDD) for observational data.
* **Granger Causality:** Use this specifically for time-series data to see if past values of X help predict Y beyond Y's own past.
* **Propensity Score Matching (PSM):** Use this to match treated and control units on confounders in observational datasets.

## Common Interview Traps
* **Trap:** Presenting a high correlation coefficient as definitive proof of causation in a business report.
* **Trap:** Confusing statistical significance (p < 0.05) with causal significance.
* **Trap:** Recommending a product intervention based on correlation (e.g., forcing users to use a specific feature because engaged users use it, ignoring that forcing it won't magically make them engaged).

## Important Formula
* **Pearson r:** `sum((xi - x_bar)(yi - y_bar)) / (n * sigma_x * sigma_y)`

## 45-Second Interview Answer
"Correlation is a necessary but not sufficient condition for causation. While correlation shows that two variables move together statistically, causation proves that one directly affects the other. In business data, correlations are often spurious, driven by hidden confounders or reverse causality. To truly establish causation, we must run Randomized Controlled Trials, or A/B tests. If experimentation isn't possible, we have to rely on causal inference techniques like Difference-in-Differences or Propensity Score Matching on observational data."

## Practice Questions:

### Q1: Shoe size and reading ability are correlated in children. What is the confounder?

**Answer:**
The confounder is age. As children grow older, their shoe size naturally increases, and they also receive more education, which improves their reading ability. If we controlled for age, the correlation between shoe size and reading ability would disappear.

**Common Mistakes Candidates Make:**
* Struggling to identify the obvious biological variable, overcomplicating the statistical explanation instead of providing the simple logical confounder.

**Likely Interviewer Follow-up:**
"How would you statistically prove that age is the confounder using this dataset?" (Answer: I would calculate the partial correlation between shoe size and reading ability while controlling for age; if the correlation drops to near zero, it confirms age is the confounding variable.)

### Q2: A data scientist shows that users who log in daily have 50% lower churn. The PM wants to add a push notification to force daily login. What is wrong with this plan?

**Answer:**
The PM is confusing correlation with causation. Forcing users to log in does not directly cause them to stop churning. The hidden confounder here is "user engagement" or "product value realization." Highly engaged users naturally log in daily AND churn less. Sending a push notification might increase the daily login metric artificially, but it won't increase underlying engagement, meaning the churn rate for those unengaged users will likely remain unchanged, or even worsen due to notification fatigue.

**Common Mistakes Candidates Make:**
* Failing to explicitly name the confounder (e.g., "engagement" or "intent").
* Agreeing with the PM that it's worth a try without highlighting the strict causal flaw first.

**Likely Interviewer Follow-up:**
"How would you design an experiment to test if the push notifications actually impact churn?" (Answer: I would set up an A/B test randomly assigning users to either receive the push notification or a holdout control group, and then measure if there is a statistically significant difference in churn rates between the two.)

### Q3: What are three methods to establish causation from observational data?

**Answer:**
When A/B testing is not possible, three common causal inference methods for observational data include:
1. **Difference-in-Differences (DiD):** Comparing the changes in outcomes over time between a group that receives an intervention and a control group that does not.
2. **Propensity Score Matching (PSM):** Matching users in a treatment group with statistically identical users in a control group based on known confounding variables.
3. **Regression Discontinuity Design (RDD):** Exploiting a strict cutoff or threshold (like a passing grade or age limit) that assigns treatment, comparing observations sitting closely on either side of the threshold.

**Common Mistakes Candidates Make:**
* Naming predictive machine learning models (like Random Forest or XGBoost) which optimize for correlation, not causation.
* Confusing observational methods with experimental design (like listing A/B testing, which contradicts the premise of the question).

**Likely Interviewer Follow-up:**
"What is the core assumption you must meet for Difference-in-Differences to be valid?" (Answer: The Parallel Trends Assumption, which states that in the absence of the intervention, the metric for both the treatment and control groups would have moved in the exact same trajectory over time.)

### Q4: What is Simpson's Paradox? Give an example.

**Answer:**
Simpson's Paradox occurs when a trend or correlation appears in several different groups of data but disappears or reverses when those groups are combined. It highlights the danger of omitted variable bias. For example, a hospital might look like it has a higher overall mortality rate than a local clinic. However, when you stratify the data by patient condition (severe vs. mild), the hospital actually has a *lower* mortality rate in both categories. The overall rate was skewed because the hospital takes in vastly more "severe" cases than the clinic. The patient's condition is the confounding variable that reverses the aggregate trend.

**Common Mistakes Candidates Make:**
* Only giving the definition without a clear, intuitive business or medical example.
* Confusing it with a standard spurious correlation. (Simpson's specifically involves the *reversal* of a trend upon aggregation/stratification).

**Likely Interviewer Follow-up:**
"In Python or SQL, how do you generally check if Simpson's Paradox is occurring in your dataset?" (Answer: I would use groupby() in Pandas or GROUP BY in SQL on suspected confounding categorical variables to verify if the directional trend of the metric reverses within the subgroups compared to the overall aggregate.)

### Q5:
**Scenario:**
You are evaluating an A/B test for two different email marketing campaigns (Campaign A and Campaign B) aiming to maximize Click-Through Rate (CTR).

Your VP of Marketing runs a simple aggregation and tells you: "Campaign B is the winner! It has a 6.5% CTR, while Campaign A only has a 5.5% CTR."

You suspect Simpson's Paradox is occurring because the campaigns were sent to two entirely different user segments: Free Tier users and Premium Tier users, in unequal distributions.

**Question:**

Write a concise Pandas code snippet to calculate the aggregate CTR for both campaigns, AND the CTR broken down by both campaign and user_tier. Based on standard Simpson's Paradox logic, what do you expect this grouped calculation to reveal to the VP?

````python
# Data:
import pandas as pd

data = {
    'campaign': ['A', 'A', 'A', 'A', 'B', 'B', 'B', 'B'],
    'user_tier': ['Premium', 'Premium', 'Free', 'Free', 'Premium', 'Premium', 'Free', 'Free'],
    'emails_sent': [1000, 1000, 9000, 9000, 9000, 9000, 1000, 1000],
    'clicks': [80, 80, 410, 420, 630, 630, 20, 20]
}
df = pd.DataFrame(data)
````

````python
# Calculate grouped CTR
grouped_df = df.groupby(['campaign', 'user_tier'])[['emails_sent', 'clicks']].sum().assign(
    CTR = lambda x: x['clicks'] / x['emails_sent']
)

# Calculate aggregate CTR
aggregate_df = df.groupby('campaign')[['emails_sent', 'clicks']].sum().assign(
    CTR = lambda x: x['clicks'] / x['emails_sent']
)
````

````python
grouped_df
````

````python
aggregate_df
````

**Interpretation:** This is a clear case of Simpson's Paradox. When we look at the grouped data, Campaign A actually has a higher CTR across *both* individual tiers (Free and Premium). However, because Campaign B sent vastly more emails to the historically higher-clicking 'Premium' tier, the aggregate metric was skewed, making Campaign B look like the overall winner.

**Common Mistakes Candidates Make:**
*   **The Average of Averages Trap:** Trying to calculate the overall CTR by running `.mean()` on a pre-existing CTR column. You cannot average ratios; you must always sum the raw numerators and denominators first before dividing.
*   Failing to articulate *why* the paradox happened (unequal sample sizes confounding the weighted average).

**Interview Tip:** 
When presenting this to a stakeholder (like a VP), never just show the code. Always lead with the business translation: "Campaign A is actually our best performer. Campaign B only looks better overall because it was heavily biased toward our easiest-to-convert users."
