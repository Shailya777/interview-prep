# Statistics Cheat Sheet

## Important Symbols
| Symbol | Meaning |
| :--- | :--- |
| $\mu$ | Population mean |
| $\sigma$ | Population standard deviation |
| $\sigma^2$ | Population variance |
| $\bar{x}$ | Sample mean |
| $s$ | Sample standard deviation |
| $s^2$ | Sample variance |
| $n$ | Sample size |
| $N$ | Population size |
| $p$ | Probability or population proportion |
| $\hat{p}$ | Sample proportion |
| $H_0$ | Null hypothesis |
| $H_1$ | Alternative hypothesis |
| $\alpha$ | Significance level (usually 0.05) |
| $z$ | Z-score / standard normal value |
| $t$ | t-statistic |

---

## Central Tendency
*   **Mean**: 
    $$\bar{x} = \frac{\sum x_i}{n}$$
*   **Median**: Middle value when data is sorted.
*   **Mode**: Most frequently occurring value.

---

## Spread / Variability
*   **Sample Variance**: 
    $$s^2 = \frac{\sum (x_i - \bar{x})^2}{n - 1}$$
*   **Standard Deviation**: 
    $$s = \sqrt{s^2}$$
*   **Interquartile Range (IQR)**: 
    $$IQR = Q_3 - Q_1$$
*   **Z-score**: 
    $$z = \frac{x - \mu}{\sigma}$$

---

## Probability
*   **Basic Probability**: 
    $$P(A) = \frac{\text{Favorable outcomes}}{\text{Total outcomes}}$$
*   **Independent Events (A and B)**: 
    $$P(A \cap B) = P(A) \times P(B)$$
*   **Mutually Exclusive / Union (A or B)**: 
    $$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$
*   **Conditional Probability**: 
    $$P(A|B) = \frac{P(A \cap B)}{P(B)}$$
*   **Bayes' Theorem**: 
    $$P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}$$

---

## Inference
*   **Standard Error (SE)**: 
    $$SE = \frac{\sigma}{\sqrt{n}}$$
*   **Confidence Interval**: 
    $$CI = \bar{x} \pm z \times \left(\frac{\sigma}{\sqrt{n}}\right)$$
*   **Z-test statistic**: 
    $$z = \frac{\bar{x} - \mu_0}{\frac{\sigma}{\sqrt{n}}}$$
*   **t-statistic**: 
    $$t = \frac{\bar{x} - \mu_0}{\frac{s}{\sqrt{n}}}$$
