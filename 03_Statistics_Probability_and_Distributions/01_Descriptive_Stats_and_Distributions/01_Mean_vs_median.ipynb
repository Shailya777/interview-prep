# Topic: Statistics: Mean vs Median

## Definition (30-second explanation)
* **Mean**: The arithmetic average, calculated by summing all values and dividing by the total count.
* **Median**: The exact middle value when a dataset is sorted; marks the 50th percentile.
* **Trimmed Mean**: A compromise metric calculated by dropping a specific percentage of the highest and lowest values before taking the average.

## Why Interviewers Ask This
* To see if you blindly apply math formulas or actively check data distribution (EDA) first.
* To evaluate your business intuition—ensuring you choose metrics that represent the "typical" user rather than letting anomalies distort reports.

## Core Concepts
* **Symmetry**: In perfectly normal, symmetric distributions, Mean ≈ Median.
* **Skewness**: Positive/Right skew (long right tail) pulls Mean > Median. Negative/Left skew pulls Mean < Median.
* **Robustness**: The median is robust to extreme outliers; the mean is highly sensitive to them.
* **Rule of Thumb**: If the absolute skewness coefficient is > 1, the median is generally the better measure of central tendency.

## When to Use
* **Mean**: For symmetric, bounded, or well-behaved continuous data without extreme outliers (e.g., Daily Active Users, Customer Age).
* **Median**: For skewed data or when you need a "typical" representation (e.g., Salaries, House Prices, Response Times).
* **Trimmed Mean**: When you want the smooth mathematical properties of a mean but need to filter out sensor noise or extreme anomalies.

## Advantages
* **Mean**: Incorporates every data point, mathematically tractable, and easy to compute at scale (e.g., MapReduce).
* **Median**: Unaffected by extreme outliers, giving a truer picture of the typical experience in skewed datasets.
* **Trimmed Mean**: Balances robustness and efficiency by discarding only the extreme tails.

## Limitations
* **Mean**: Highly susceptible to extreme values (one massive outlier ruins the metric).
* **Median**: Computationally expensive at scale because it requires sorting the data ($O(n \log n)$ complexity); ignores the magnitude of valid extreme events.

## Common Comparisons
* **Mean vs. Median**: Mean shifts easily with outliers; Median holds steady based on rank position. 
* **Median vs. Trimmed Mean**: Median practically discards all values except the middle 1-2 points. Trimmed mean discards only the extreme X% and incorporates the rest.

## Common Interview Traps
* **Trap 1:** Blindly recommending the mean without stating: "I would first visualize the data to check for outliers."
* **Trap 2:** Recommending the median for perfectly symmetric data (the mean is statistically more efficient/informative here).
* **Trap 3:** Forgetting that finding a median requires sorting data, which is a common trick in system design or algorithmic rounds.
* **Trap 4:** Forgetting to average the two middle numbers when the sample size ($n$) is even.

## Important Formula (if applicable)

* **Mean**: $\frac{1}{n} \sum_{i=1}^{n} x_i$
* **Median**: $X_{(n+1)/2}$ (if $n$ is odd), or $\frac{X_{n/2} + X_{(n/2)+1}}{2}$ (if $n$ is even)

## 45-Second Interview Answer
"Choosing between the mean and median entirely depends on the underlying data distribution. I always start by visualizing the data via a histogram or checking its skewness. If the data is roughly symmetric without extreme outliers, the mean is ideal because it uses all data points. However, for metrics like salaries or web response times, which naturally have a long right tail of extreme values, the mean becomes inflated. In those cases, the median, or a trimmed mean, is much better for reporting the 'typical' observation to stakeholders."

## Example Questions:

### Question 1
Question: A dataset of house prices has skewness = 2.3. Which measure of center would you report and why?

Ideal Interview Answer: 
"I would report the median. A skewness coefficient of 2.3 indicates a heavily right-skewed distribution, meaning there is a long tail of very expensive luxury homes. Because the absolute skewness is greater than 1, the mean would be artificially inflated by those luxury outliers. The median will provide a much more accurate representation of a 'typical' house price."
(MySQL Note: If answering with data, you would write a query to extract the 50th percentile, as simple averages would be misleading here.)

Common Mistakes Candidates Make: 
- Saying "Mean" because they ignore the skewness coefficient threshold.
- Failing to explain *why* the skewness pulls the mean upward.

One Likely Interviewer Follow-Up: 
"If you still needed to report an 'average' because executives prefer that term, how could you manipulate the data to make the mean more representative?" (Answer: Use a trimmed mean to drop the top and bottom 5-10%).

### Question 2
Question: Five friends have incomes: $30K, $32K, $35K, $28K, $500K. Calculate both mean and median.

Ideal Interview Answer: 
"First, to calculate the Mean: (30 + 32 + 35 + 28 + 500) / 5 = 625 / 5 = $125K.
Second, to calculate the Median, I must sort the data: 28, 30, 32, 35, 500. Since n=5 (odd), the median is the 3rd value, which is $32K.
This perfectly highlights why median is better here; nobody actually makes the 'average' of $125K."

Common Mistakes Candidates Make: 
- Forgetting to sort the values before finding the median.
- Calculation errors under pressure.

One Likely Interviewer Follow-Up: 
"Write the SQL logic to find this average excluding the maximum value."

### Question 3
Question: When would you use both mean AND median together in a business report?

Ideal Interview Answer: 
"I would report both side-by-side to highlight the presence of extreme outliers or heavy skewness to stakeholders. For example, if I report that the Mean Customer LTV is $500 but the Median LTV is $50, it immediately tells the business that while the 'typical' customer spends $50, we have a small segment of 'whale' customers massively driving up overall revenue. It gives a complete picture of both typical behavior and total scale."

Common Mistakes Candidates Make: 
- Thinking you can only ever choose one metric.
- Failing to tie the statistical difference to a business insight (like identifying "whales").

One Likely Interviewer Follow-Up: 
"If Mean > Median, what does that tell you about the shape of the distribution?" (Answer: Right-skewed).

### Question 4
Question: How does adding one extreme outlier affect the mean versus the median?

Ideal Interview Answer: 
"Adding one extreme outlier will drastically pull the mean in the direction of the outlier because the mean sums all values. The median, however, is robust. Adding one data point simply shifts the middle rank by at most half a step (e.g., from one specific value to the average of two middle values), leaving the median virtually unchanged."

Common Mistakes Candidates Make: 
- Saying the median "doesn't change at all" (it might shift slightly if it changes an odd *n* to an even *n*).

One Likely Interviewer Follow-Up: 
"What happens to the standard deviation when that extreme outlier is added?" (Answer: It increases significantly, as standard deviation squares the distance from the mean).

## Practice Questions:

### Q1: MySQL
**You are given a table of employee salaries. MySQL does not have a built-in MEDIAN() function. Write a MySQL query to calculate the exact median salary for all employees.**

***Mock Schema**
```sql
CREATE TABLE salaries (
    emp_id INT,
    salary INT
);

INSERT INTO salaries (emp_id, salary) VALUES 
(1, 28000), 
(2, 30000), 
(3, 32000), 
(4, 35000), 
(5, 500000);
```

**Answer**
```sql
-- Option 1:
WITH ranked_salaries AS (
    SELECT 
        emp_id, 
        salary,
        -- Added emp_id as a tie-breaker for deterministic sorting
        ROW_NUMBER() OVER(ORDER BY salary ASC, emp_id ASC) as rnk,
        COUNT(*) OVER() as total_rows
    FROM salaries
)
SELECT AVG(salary) AS median
FROM ranked_salaries
WHERE rnk BETWEEN total_rows / 2.0 AND total_rows / 2.0 + 1;


-- Option 2:
with ranked_salaries as (
	select emp_id, salary,
  	row_number() over(order by salary asc, emp_id) as rnk_asc,
  	row_number() over(order by salary desc, emp_id) as rnk_desc
  from salaries
)
select avg(salary) as median
from ranked_salaries
where abs(cast(rnk_asc as signed) - cast(rnk_desc as signed)) <= 1;
```

**Interview Tips for SQL Medians**
- The "Odd/Even" Trap: The mathematical bound BETWEEN total_rows / 2.0 AND total_rows / 2.0 + 1 flawlessly captures the single middle row if the count is odd, and the two middle rows if the count is even. AVG() then resolves the final number.

- The "Deterministic" Trap: Always add a unique identifier (like id or emp_id) to your ORDER BY clause when using window functions like ROW_NUMBER(). Otherwise, duplicate values can cause unpredictable row numbering.

- The "Unsigned Integer" Trap: If using the ascending/descending rank difference method, you must cast the ranks to SIGNED in MySQL before subtracting them, or you will throw an out-of-range error.

- Efficiency: Sorting a dataset twice (Ascending and Descending) is expensive. Using COUNT(*) OVER() allows you to sort the data only once.

### Q2: Statistics: Mean vs Median (Big Data & Streaming)

## Question: The Streaming Median Problem
**Question:** How do you calculate a rolling median for a real-time data stream of 100,000 requests per second?

## Ideal Interview Answer
"Calculating an *exact* median on a high-throughput stream is generally an anti-pattern. Because the median requires sorting, it has an $O(n \log n)$ time complexity and requires all data points to be stored in memory ($O(n)$ space). At 100k requests per second, the system will instantly bottleneck. 

To solve this in production, I would use an **Approximate Percentile algorithm**, specifically **t-digest** or an **HDR Histogram**. These probabilistic data structures allow us to estimate the median and P99s with high accuracy (e.g., 99.9%) using a fixed, tiny memory footprint ($O(1)$ space and time). If approximation is not allowed, I would propose micro-batching the data into 5-minute intervals, calculating the exact median offline, and accepting the delay."

## Interview Tips for Big Data Medians
* **The $O(n \log n)$ Trap:** Interviewers ask this specifically to see if you know that medians cannot be easily distributed or calculated on the fly like means (which just require a rolling `sum / count`).
* **The Magic Keyword:** "t-digest" (used under the hood by Elasticsearch, Redis, and Apache Spark for percentile calculations).
* **Alternative approach - Histograms/Bucketing:** If you don't know t-digest, suggest creating fixed buckets (e.g., 0-10ms, 11-20ms, 21-50ms) and simply incrementing a counter when a request falls in that bucket. You can then interpolate the median from the bucket counts in real-time.

### Q3:
**You are analyzing text sequence lengths (in tokens) for a new NLP training corpus. Most sentences are standard length, but a few massive documents have slipped into the dataset.**
**Using the provided mock dataset, write a Python snippet that does the following:**

- Loads the data into a Pandas DataFrame and prints the standard summary statistics.
- Calculates and prints the exact Mean and Median using NumPy.
- Calculates and prints the skewness of the dataset and a 10% Trimmed Mean using scipy.stats.

````python
# Data:
# Token lengths of 10 sample sequences in the NLP corpus
token_lengths = [12, 15, 18, 14, 22, 19, 16, 25, 20, 1500]
````

````python
import pandas as pd
import numpy as np
from scipy.stats import skew, trim_mean

# 1. Load data and print standard summary statistics
token_lengths = [12, 15, 18, 14, 22, 19, 16, 25, 20, 1500]
df = pd.DataFrame({'Token Lengths': token_lengths})
print("--- Pandas Summary Statistics ---")
print(df.describe())
print("\n")

# 2. Calculate Exact Mean and Median (NumPy)
# The mean will be heavily skewed by the 1500 token document
mean_val = np.mean(token_lengths)
median_val = np.median(token_lengths)
print(f"NumPy Mean: {mean_val}")
print(f"NumPy Median: {median_val}")

# 3. Calculate Skewness and Trimmed Mean (SciPy)
# Dropping the top and bottom 10% (1 value from each end for n=10)
skewness = skew(token_lengths)
t_mean = trim_mean(token_lengths, proportiontocut=0.1)

print(f"Skewness: {skewness:.2f} (Highly right-skewed)")
print(f"Trimmed Mean (10%): {t_mean:.2f}")
````

**Interview Tips**
- Pandas Describe: Always use df.describe() when asked for a quick data summary. It instantly provides count, mean, std, min, 25%, 50% (median), 75%, and max.
- Trimmed Mean utility: In machine learning data prep (like cleaning NLP corpora), a trimmed mean is highly practical for dealing with scraping errors or corrupted files without throwing away mostly good data.
